```python
# ============================================================
# INTEGRATED PAYMENT SYSTEM DATA ANALYSIS
# Comparative Analysis of 3 Cleaned Datasets
# ============================================================
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully.")
```

```python
# ============================================================
# LOAD THE THREE CLEANED DATASETS
# ============================================================

import pandas as pd

file1 = "NPCI_Final_Cleaned.xlsx"
file2 = "cleaned_atm_2025.csv"
file3 = "PSI_cleaned.xlsx"

# Load datasets
df1 = pd.read_excel(file1)
df2 = pd.read_csv(file2)
df3 = pd.read_excel(file3)

print("=" * 80)
print("THREE CLEANED DATASETS LOADED SUCCESSFULLY")
print("=" * 80)

print("\nDATASET 1 - NPCI")
print("Rows    :", df1.shape[0])
print("Columns :", df1.shape[1])

print("\nDATASET 2 - ATM")
print("Rows    :", df2.shape[0])
print("Columns :", df2.shape[1])

print("\nDATASET 3 - PSI")
print("Rows    :", df3.shape[0])
print("Columns :", df3.shape[1])

print("\n" + "=" * 80)
print("ALL THREE DATASETS LOADED SUCCESSFULLY")
print("=" * 80)
```

```python
# ============================================================
# STEP 2 - STRUCTURE & COLUMN ANALYSIS OF ALL THREE DATASETS
# ============================================================

datasets = {
    "NPCI": df1,
    "ATM": df2,
    "PSI": df3
}

for name, df in datasets.items():

    print("\n" + "=" * 90)
    print(f"{name} DATASET")
    print("=" * 90)

    print("\nShape:")
    print("Rows    :", df.shape[0])
    print("Columns :", df.shape[1])

    print("\nColumn Names:")
    for i, col in enumerate(df.columns, 1):
        print(f"{i:02d}. {col}")

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum()[df.isnull().sum() > 0])

    print("\nFirst 3 Rows:")
    display(df.head(3))

print("\n" + "=" * 90)
print("STRUCTURE ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# STEP 3 - COMMON TIME PERIOD ANALYSIS
# ============================================================

print("=" * 90)
print("COMMON TIME PERIOD ANALYSIS")
print("=" * 90)

# ------------------------------------------------------------
# ATM DATASET
# ------------------------------------------------------------
print("\nATM DATASET")
print("-" * 50)

print("Years:", sorted(df2["Year"].unique()))
print("Months:", df2["Month"].unique().tolist())
print("Quarters:", df2["Quarter"].unique().tolist())

print("\nATM records by Year:")
print(df2.groupby("Year").size())


# ------------------------------------------------------------
# PSI DATASET
# ------------------------------------------------------------
print("\n" + "=" * 90)
print("PSI DATASET")
print("-" * 50)

print("Years:", sorted(df3["Year"].unique()))
print("Months:", df3["Month"].unique().tolist())
print("Quarters:", df3["Quarter"].unique().tolist())

print("\nPSI records by Year:")
print(df3.groupby("Year").size())


# ------------------------------------------------------------
# NPCI DATASET
# ------------------------------------------------------------
print("\n" + "=" * 90)
print("NPCI DATASET")
print("-" * 50)

# Show the transaction period columns
npci_period_columns = [
    col for col in df1.columns
    if "Volume" in col or "Value" in col
]

for col in npci_period_columns:
    print(col)

print("\n" + "=" * 90)
print("COMMON ANALYSIS PERIOD")
print("=" * 90)

print("""
ATM and PSI contain explicit Year/Month/Quarter fields.
NPCI contains quarterly FY periods plus April-June 2026.

We will identify the overlapping period before combining
the datasets for comparative analysis.
""")
```

```python
# ============================================================
# STEP 3 - IDENTIFY COMMON ANALYSIS PERIODS
# ============================================================

print("=" * 90)
print("COMMON QUARTERLY ANALYSIS PERIODS")
print("=" * 90)

# ATM and PSI: calendar-year 2025 quarters
common_periods = {
    "2025_Q2": {
        "ATM_Quarter": "Q2",
        "PSI_Quarter": "Q2",
        "NPCI_Period": "FY_2025_2026_Q1"
    },
    "2025_Q3": {
        "ATM_Quarter": "Q3",
        "PSI_Quarter": "Q3",
        "NPCI_Period": "FY_2025_2026_Q2"
    },
    "2025_Q4": {
        "ATM_Quarter": "Q4",
        "PSI_Quarter": "Q4",
        "NPCI_Period": "FY_2025_2026_Q3"
    }
}

for period, mapping in common_periods.items():
    print(f"\n{period}")
    print("-" * 50)
    print("ATM Quarter  :", mapping["ATM_Quarter"])
    print("PSI Quarter  :", mapping["PSI_Quarter"])
    print("NPCI Period  :", mapping["NPCI_Period"])

print("\n" + "=" * 90)
print("COMMON PERIOD SELECTION COMPLETED")
print("=" * 90)

print("""
Reason:
ATM and PSI contain calendar-year monthly data for 2025,
while NPCI contains FY-based quarterly data.

Therefore, the common comparable periods are:
Apr-Jun 2025  -> NPCI FY 2025-26 Q1
Jul-Sep 2025  -> NPCI FY 2025-26 Q2
Oct-Dec 2025  -> NPCI FY 2025-26 Q3
""")
```

```python
# ============================================================
# STEP 4 - EXTRACT COMMON ANALYSIS PERIODS
# ============================================================

print("=" * 90)
print("EXTRACTING COMMON PERIOD DATA FROM ALL THREE DATASETS")
print("=" * 90)

# ------------------------------------------------------------
# 1. ATM - 2025 Q2, Q3, Q4
# ------------------------------------------------------------

atm_common = df2[
    (df2["Year"] == 2025) &
    (df2["Quarter"].isin(["Q2", "Q3", "Q4"]))
].copy()

# ------------------------------------------------------------
# 2. PSI - 2025 Q2, Q3, Q4
# ------------------------------------------------------------

psi_common = df3[
    (df3["Year"] == 2025) &
    (df3["Quarter"].isin(["Q2", "Q3", "Q4"]))
].copy()

# ------------------------------------------------------------
# 3. NPCI - FY 2025-26 Q1, Q2, Q3
# ------------------------------------------------------------

npci_common_columns = [
    "NPCI_Operated_System",
    
    "FY_2025_2026_Q1_Volume_in_Mn",
    "FY_2025_2026_Q1_Value_in_Bn",
    
    "FY_2025_2026_Q2_Volume_in_Mn",
    "FY_2025_2026_Q2_Value_in_Bn",
    
    "FY_2025_2026_Q3_Volume_in_Mn",
    "FY_2025_2026_Q3_Value_in_Bn"
]

npci_common = df1[npci_common_columns].copy()

# Convert NPCI transaction columns to numeric
npci_numeric_columns = npci_common.columns[1:]

for col in npci_numeric_columns:
    npci_common[col] = pd.to_numeric(
        npci_common[col],
        errors="coerce"
    )

# Remove structural/non-system rows
npci_common = npci_common[
    npci_common["NPCI_Operated_System"].notna()
]

npci_common = npci_common[
    npci_common["NPCI_Operated_System"] != "Financial Txns:"
]

# ------------------------------------------------------------
# DISPLAY RESULTS
# ------------------------------------------------------------

print("\nATM COMMON DATA")
print("-" * 60)
print("Rows    :", atm_common.shape[0])
print("Columns :", atm_common.shape[1])
print("Quarters:", sorted(atm_common["Quarter"].unique()))

print("\nPSI COMMON DATA")
print("-" * 60)
print("Rows    :", psi_common.shape[0])
print("Columns :", psi_common.shape[1])
print("Quarters:", sorted(psi_common["Quarter"].unique()))

print("\nNPCI COMMON DATA")
print("-" * 60)
print("Rows    :", npci_common.shape[0])
print("Columns :", npci_common.shape[1])

print("\nNPCI Systems:")
print(npci_common["NPCI_Operated_System"].tolist())

print("\n" + "=" * 90)
print("COMMON PERIOD DATA EXTRACTION COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# STEP 5 - CREATE QUARTERLY SUMMARY TABLES
# ============================================================

print("=" * 90)
print("CREATING COMPARABLE QUARTERLY SUMMARY TABLES")
print("=" * 90)

# ============================================================
# A. ATM QUARTERLY SUMMARY
# ============================================================

atm_quarterly = atm_common.groupby(
    "Quarter"
).agg(
    Total_ATM_Onsite=("Infrastructure - ATMs & CRMs - On-site", "sum"),
    Total_ATM_Offsite=("Infrastructure - ATMs & CRMs - Off-site", "sum"),
    Total_POS=("Infrastructure - PoS", "sum"),
    Total_Micro_ATMs=("Infrastructure - Micro ATMs", "sum"),
    Total_Bharat_QR=("Infrastructure - Bharat QR Codes", "sum"),
    Total_UPI_QR=("Infrastructure - UPI QR Codes", "sum"),
    Total_Credit_Cards=("Infrastructure - Credit Cards", "sum"),
    Total_Debit_Cards=("Infrastructure - Debit Cards", "sum"),
    CreditCard_POS_Volume=(
        "Credit Card - Card Payments - PoS - Volume (in actuals)", "sum"
    ),
    CreditCard_Online_Volume=(
        "Credit Card - Card Payments - Online (e-com) - Volume (in actuals)", "sum"
    ),
    DebitCard_POS_Volume=(
        "Debit Card - Card Payments - PoS - Volume (in actuals)", "sum"
    ),
    DebitCard_Online_Volume=(
        "Debit Card - Card Payments - Online (e-com) - Volume (in actuals)", "sum"
    ),
    DebitCard_ATM_Withdrawal_Volume=(
        "Debit Card - Cash Withdrawal - ATM - Volume (in actuals)", "sum"
    )
).reset_index()


# ============================================================
# B. PSI QUARTERLY SUMMARY
# ============================================================

psi_quarterly = psi_common.groupby(
    "Quarter"
).agg(
    PSI_Volume_Current_Month=(
        "Volume (lakh) - Current Month", "sum"
    ),
    PSI_Volume_Previous_Month=(
        "Volume (lakh) - Previous Month", "sum"
    ),
    PSI_Volume_Same_Month_Last_Year=(
        "Volume (lakh) - Same Month Last Year", "sum"
    ),
    PSI_Value_Current_Month=(
        "Value (₹ crore) - Current Month", "sum"
    ),
    PSI_Value_Previous_Month=(
        "Value (₹ crore) - Previous Month", "sum"
    ),
    PSI_Value_Same_Month_Last_Year=(
        "Value (₹ crore) - Same Month Last Year", "sum"
    )
).reset_index()


# ============================================================
# C. NPCI QUARTERLY SUMMARY
# ============================================================

npci_quarterly = pd.DataFrame({
    "Quarter": ["Q2", "Q3", "Q4"],

    "NPCI_Volume_Mn": [
        npci_common["FY_2025_2026_Q1_Volume_in_Mn"].sum(),
        npci_common["FY_2025_2026_Q2_Volume_in_Mn"].sum(),
        npci_common["FY_2025_2026_Q3_Volume_in_Mn"].sum()
    ],

    "NPCI_Value_Bn": [
        npci_common["FY_2025_2026_Q1_Value_in_Bn"].sum(),
        npci_common["FY_2025_2026_Q2_Value_in_Bn"].sum(),
        npci_common["FY_2025_2026_Q3_Value_in_Bn"].sum()
    ]
})


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("\nATM QUARTERLY SUMMARY")
print("-" * 90)
display(atm_quarterly)

print("\nPSI QUARTERLY SUMMARY")
print("-" * 90)
display(psi_quarterly)

print("\nNPCI QUARTERLY SUMMARY")
print("-" * 90)
display(npci_quarterly)

print("\n" + "=" * 90)
print("QUARTERLY SUMMARY TABLES CREATED SUCCESSFULLY")
print("=" * 90)
```

```python
# ============================================================
# STEP 6 - COMBINE THREE DATASETS
# ============================================================

print("=" * 90)
print("COMBINING NPCI, ATM AND PSI QUARTERLY SUMMARIES")
print("=" * 90)

# Merge ATM and PSI
combined_analysis = pd.merge(
    atm_quarterly,
    psi_quarterly,
    on="Quarter",
    how="inner"
)

# Merge NPCI
combined_analysis = pd.merge(
    combined_analysis,
    npci_quarterly,
    on="Quarter",
    how="inner"
)

# Add readable period labels
period_mapping = {
    "Q2": "Apr-Jun 2025",
    "Q3": "Jul-Sep 2025",
    "Q4": "Oct-Dec 2025"
}

combined_analysis["Analysis_Period"] = (
    combined_analysis["Quarter"].map(period_mapping)
)

# Put period first
cols = ["Analysis_Period", "Quarter"] + [
    col for col in combined_analysis.columns
    if col not in ["Analysis_Period", "Quarter"]
]

combined_analysis = combined_analysis[cols]

# Reorder chronologically
quarter_order = ["Q2", "Q3", "Q4"]

combined_analysis["Quarter"] = pd.Categorical(
    combined_analysis["Quarter"],
    categories=quarter_order,
    ordered=True
)

combined_analysis = combined_analysis.sort_values(
    "Quarter"
).reset_index(drop=True)

# Display
print("\nCOMBINED ANALYTICAL DATASET")
print("-" * 90)

print("Rows    :", combined_analysis.shape[0])
print("Columns :", combined_analysis.shape[1])

display(combined_analysis)

print("\nMissing Values:")
print(
    combined_analysis.isnull().sum()[
        combined_analysis.isnull().sum() > 0
    ]
)

print("\n" + "=" * 90)
print("THREE DATASETS COMBINED SUCCESSFULLY")
print("=" * 90)
```

```python
# ============================================================
# STEP 7 - QUARTER-OVER-QUARTER GROWTH ANALYSIS
# ============================================================

print("=" * 90)
print("QUARTER-OVER-QUARTER GROWTH ANALYSIS")
print("=" * 90)

# Important indicators selected for cross-dataset analysis
growth_columns = [
    "Total_ATM_Onsite",
    "Total_ATM_Offsite",
    "Total_POS",
    "Total_Micro_ATMs",
    "Total_Bharat_QR",
    "Total_UPI_QR",
    "Total_Credit_Cards",
    "Total_Debit_Cards",
    "CreditCard_POS_Volume",
    "CreditCard_Online_Volume",
    "DebitCard_POS_Volume",
    "DebitCard_Online_Volume",
    "DebitCard_ATM_Withdrawal_Volume",
    "PSI_Volume_Current_Month",
    "PSI_Value_Current_Month",
    "NPCI_Volume_Mn",
    "NPCI_Value_Bn"
]

# Create growth table
growth_analysis = combined_analysis[
    ["Analysis_Period", "Quarter"] + growth_columns
].copy()

# Calculate percentage change from previous quarter
for col in growth_columns:
    growth_analysis[col + "_Growth_%"] = (
        growth_analysis[col]
        .pct_change() * 100
    )

# Display only growth columns
growth_columns_display = [
    "Analysis_Period",
    "Quarter"
] + [
    col + "_Growth_%"
    for col in growth_columns
]

growth_result = growth_analysis[growth_columns_display].copy()

print("\nQUARTER-OVER-QUARTER GROWTH (%)")
print("-" * 90)

display(growth_result)

print("\n" + "=" * 90)
print("GROWTH ANALYSIS COMPLETED")
print("=" * 90)

print("""
Interpretation:
• Q2 is the baseline, so its growth values are NaN.
• Q3 growth represents change from Q2 to Q3.
• Q4 growth represents change from Q3 to Q4.
• Positive values indicate growth.
• Negative values indicate decline.
""")
```

```python
# ============================================================
# CHECK AVAILABLE DATAFRAME VARIABLES
# ============================================================

print("=" * 90)
print("AVAILABLE DATAFRAME VARIABLES")
print("=" * 90)

dataframe_list = [
    (name, obj)
    for name, obj in list(globals().items())
    if isinstance(obj, pd.DataFrame)
]

for name, obj in dataframe_list:
    print(f"{name:25s} -> {obj.shape[0]} rows × {obj.shape[1]} columns")

print("=" * 90)
```

```python
# ============================================================
# STEP - CROSS-DATASET CORRELATION ANALYSIS
# ============================================================

print("=" * 90)
print("CROSS-DATASET CORRELATION ANALYSIS")
print("=" * 90)

# Important variables from the combined dataset
correlation_columns = [
    'Total_ATM_Onsite',
    'Total_ATM_Offsite',
    'Total_POS',
    'Total_Micro_ATMs',
    'Total_Bharat_QR',
    'Total_UPI_QR',
    'Total_Credit_Cards',
    'Total_Debit_Cards',
    'CreditCard_POS_Volume',
    'CreditCard_Online_Volume',
    'DebitCard_POS_Volume',
    'DebitCard_Online_Volume',
    'DebitCard_ATM_Withdrawal_Volume',
    'PSI_Volume_Current_Month',
    'PSI_Value_Current_Month',
    'NPCI_Volume_Mn',
    'NPCI_Value_Bn'
]

# Calculate correlation matrix
corr_matrix = combined_analysis[correlation_columns].corr()

print("\nCORRELATION MATRIX")
print("-" * 90)

display(corr_matrix.round(3))

print("\n" + "=" * 90)
print("CORRELATION WITH NPCI TRANSACTION VOLUME")
print("-" * 90)

npci_volume_corr = (
    corr_matrix['NPCI_Volume_Mn']
    .drop('NPCI_Volume_Mn')
    .sort_values(key=abs, ascending=False)
)

display(
    npci_volume_corr
    .to_frame("Correlation_with_NPCI_Volume")
    .round(3)
)

print("\n" + "=" * 90)
print("CORRELATION WITH NPCI TRANSACTION VALUE")
print("-" * 90)

npci_value_corr = (
    corr_matrix['NPCI_Value_Bn']
    .drop('NPCI_Value_Bn')
    .sort_values(key=abs, ascending=False)
)

display(
    npci_value_corr
    .to_frame("Correlation_with_NPCI_Value")
    .round(3)
)

print("\n" + "=" * 90)
print("CROSS-DATASET CORRELATION ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# STEP - DERIVED KPI ANALYSIS
# ============================================================

print("=" * 90)
print("DERIVED KPI ANALYSIS")
print("=" * 90)

kpi = combined_analysis.copy()

# ------------------------------------------------------------
# 1. NPCI AVERAGE TRANSACTION VALUE
# ------------------------------------------------------------
# NPCI Volume is in million transactions
# NPCI Value is in billion currency units
# Average transaction value = (Value / Volume) * 1000

kpi['NPCI_Avg_Transaction_Value'] = (
    kpi['NPCI_Value_Bn'] / kpi['NPCI_Volume_Mn']
) * 1000


# ------------------------------------------------------------
# 2. PSI MONTH-OVER-MONTH VOLUME CHANGE
# ------------------------------------------------------------
kpi['PSI_Volume_MoM_Change_%'] = (
    (kpi['PSI_Volume_Current_Month'] -
     kpi['PSI_Volume_Previous_Month'])
    / kpi['PSI_Volume_Previous_Month']
) * 100


# ------------------------------------------------------------
# 3. PSI MONTH-OVER-MONTH VALUE CHANGE
# ------------------------------------------------------------
kpi['PSI_Value_MoM_Change_%'] = (
    (kpi['PSI_Value_Current_Month'] -
     kpi['PSI_Value_Previous_Month'])
    / kpi['PSI_Value_Previous_Month']
) * 100


# ------------------------------------------------------------
# 4. CARD PAYMENT VOLUME
# ------------------------------------------------------------
kpi['Total_Card_Payment_Volume'] = (
    kpi['CreditCard_POS_Volume'] +
    kpi['CreditCard_Online_Volume'] +
    kpi['DebitCard_POS_Volume'] +
    kpi['DebitCard_Online_Volume']
)


# ------------------------------------------------------------
# 5. DIGITAL QR INFRASTRUCTURE
# ------------------------------------------------------------
kpi['Total_QR_Infrastructure'] = (
    kpi['Total_Bharat_QR'] +
    kpi['Total_UPI_QR']
)


# ------------------------------------------------------------
# 6. ATM INFRASTRUCTURE
# ------------------------------------------------------------
kpi['Total_ATM_Infrastructure'] = (
    kpi['Total_ATM_Onsite'] +
    kpi['Total_ATM_Offsite']
)


# ------------------------------------------------------------
# DISPLAY KPI TABLE
# ------------------------------------------------------------

kpi_columns = [
    'Analysis_Period',
    'Quarter',
    'NPCI_Volume_Mn',
    'NPCI_Value_Bn',
    'NPCI_Avg_Transaction_Value',
    'PSI_Volume_Current_Month',
    'PSI_Volume_MoM_Change_%',
    'PSI_Value_Current_Month',
    'PSI_Value_MoM_Change_%',
    'Total_Card_Payment_Volume',
    'Total_QR_Infrastructure',
    'Total_ATM_Infrastructure'
]

print("\nDERIVED KPI TABLE")
print("-" * 90)

display(kpi[kpi_columns].round(3))

print("\n" + "=" * 90)
print("DERIVED KPI ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# STEP - REBUILD COMMON PERIODS FOR 4-QUARTER ANALYSIS
# ============================================================

print("=" * 90)
print("REBUILDING COMMON PERIOD DATA - 4 QUARTERS")
print("=" * 90)

# ATM and PSI: Calendar Year 2025
# Q1 = Jan-Mar, Q2 = Apr-Jun, Q3 = Jul-Sep, Q4 = Oct-Dec

atm_common_4q = df2[df2['Quarter'].isin(['Q1', 'Q2', 'Q3', 'Q4'])].copy()
psi_common_4q = df3[df3['Quarter'].isin(['Q1', 'Q2', 'Q3', 'Q4'])].copy()

# NPCI:
# Jan-Mar 2025 = FY 2024-2025 Q4
# Apr-Jun 2025 = FY 2025-2026 Q1
# Jul-Sep 2025 = FY 2025-2026 Q2
# Oct-Dec 2025 = FY 2025-2026 Q3

npci_period_columns = [
    'FY_2024_2025_Q4_Volume_in_Mn',
    'FY_2024_2025_Q4_Value_in_Bn',
    'FY_2025_2026_Q1_Volume_in_Mn',
    'FY_2025_2026_Q1_Value_in_Bn',
    'FY_2025_2026_Q2_Volume_in_Mn',
    'FY_2025_2026_Q2_Value_in_Bn',
    'FY_2025_2026_Q3_Volume_in_Mn',
    'FY_2025_2026_Q3_Value_in_Bn'
]

npci_common_4q = df1[
    ['NPCI_Operated_System'] + npci_period_columns
].copy()

print("\nATM COMMON DATA")
print("-" * 60)
print("Rows    :", len(atm_common_4q))
print("Quarters:", sorted(atm_common_4q['Quarter'].unique()))

print("\nPSI COMMON DATA")
print("-" * 60)
print("Rows    :", len(psi_common_4q))
print("Quarters:", sorted(psi_common_4q['Quarter'].unique()))

print("\nNPCI COMMON DATA")
print("-" * 60)
print("Rows    :", len(npci_common_4q))
print("Columns :", len(npci_common_4q.columns))

print("\n" + "=" * 90)
print("4-QUARTER COMMON PERIOD EXTRACTION COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# STEP - CREATE 4-QUARTER NPCI SUMMARY
# ============================================================

print("=" * 90)
print("CREATING 4-QUARTER NPCI SUMMARY")
print("=" * 90)

# Convert NPCI numeric columns to numeric
for col in npci_period_columns:
    npci_common_4q[col] = pd.to_numeric(
        npci_common_4q[col],
        errors='coerce'
    )

# Define NPCI quarter mapping
npci_quarter_mapping = {
    'Q1': ('FY_2024_2025_Q4_Volume_in_Mn',
           'FY_2024_2025_Q4_Value_in_Bn'),

    'Q2': ('FY_2025_2026_Q1_Volume_in_Mn',
           'FY_2025_2026_Q1_Value_in_Bn'),

    'Q3': ('FY_2025_2026_Q2_Volume_in_Mn',
           'FY_2025_2026_Q2_Value_in_Bn'),

    'Q4': ('FY_2025_2026_Q3_Volume_in_Mn',
           'FY_2025_2026_Q3_Value_in_Bn')
}

npci_summary_list = []

for quarter, (volume_col, value_col) in npci_quarter_mapping.items():

    volume = npci_common_4q[volume_col].sum()
    value = npci_common_4q[value_col].sum()

    npci_summary_list.append({
        'Quarter': quarter,
        'NPCI_Volume_Mn': volume,
        'NPCI_Value_Bn': value
    })

npci_quarterly_4q = pd.DataFrame(npci_summary_list)

# Add actual calendar period
period_mapping = {
    'Q1': 'Jan-Mar 2025',
    'Q2': 'Apr-Jun 2025',
    'Q3': 'Jul-Sep 2025',
    'Q4': 'Oct-Dec 2025'
}

npci_quarterly_4q['Analysis_Period'] = (
    npci_quarterly_4q['Quarter'].map(period_mapping)
)

# Arrange columns
npci_quarterly_4q = npci_quarterly_4q[
    ['Analysis_Period', 'Quarter',
     'NPCI_Volume_Mn', 'NPCI_Value_Bn']
]

print("\nNPCI 4-QUARTER SUMMARY")
print("-" * 90)

display(npci_quarterly_4q)

print("\n" + "=" * 90)
print("4-QUARTER NPCI SUMMARY CREATED SUCCESSFULLY")
print("=" * 90)
```

```python
# ============================================================
# 4-QUARTER QoQ GROWTH ANALYSIS
# ============================================================

print("=" * 90)
print("4-QUARTER QUARTER-OVER-QUARTER GROWTH ANALYSIS")
print("=" * 90)

# Work on a copy
growth_4q = combined_4q.copy()

# Numeric columns for QoQ calculation
growth_columns = [
    c for c in growth_4q.columns
    if c not in ["Analysis_Period", "Quarter"]
]

# Calculate QoQ growth %
for col in growth_columns:
    growth_4q[col + "_Growth_%"] = (
        growth_4q[col].pct_change() * 100
    )

# Keep only identification + growth columns
growth_result_4q = growth_4q[
    ["Analysis_Period", "Quarter"] +
    [c for c in growth_4q.columns if c.endswith("_Growth_%")]
].copy()

# Display
print("\n4-QUARTER QoQ GROWTH TABLE")
print("-" * 90)

display(growth_result_4q)

print("\n" + "=" * 90)
print("4-QUARTER QoQ GROWTH ANALYSIS COMPLETED")
print("=" * 90)

print("""
Interpretation:
• Q1 is the baseline, so its growth values are NaN.
• Q2 shows growth/decline compared with Q1.
• Q3 shows growth/decline compared with Q2.
• Q4 shows growth/decline compared with Q3.
• Positive values indicate quarter-over-quarter growth.
• Negative values indicate quarter-over-quarter decline.
""")
```

```python
# ============================================================
# DERIVED KPI ANALYSIS - 4 QUARTERS
# ============================================================

print("=" * 90)
print("CREATING DERIVED KPI ANALYSIS - 4 QUARTERS")
print("=" * 90)

kpi_4q = combined_4q.copy()

# ------------------------------------------------------------
# 1. NPCI Average Transaction Value
# ------------------------------------------------------------
# NPCI Value is in Billion Rs
# NPCI Volume is in Million transactions
# Average transaction value in Rs:
# (Value in Bn / Volume in Mn) * 1000

kpi_4q["NPCI_Avg_Transaction_Value"] = (
    kpi_4q["NPCI_Value_Bn"] /
    kpi_4q["NPCI_Volume_Mn"]
) * 1000


# ------------------------------------------------------------
# 2. PSI Month-over-Month Change
# ------------------------------------------------------------

kpi_4q["PSI_Volume_MoM_Change_%"] = (
    (
        kpi_4q["PSI_Volume_Current_Month"] /
        kpi_4q["PSI_Volume_Previous_Month"]
    ) - 1
) * 100

kpi_4q["PSI_Value_MoM_Change_%"] = (
    (
        kpi_4q["PSI_Value_Current_Month"] /
        kpi_4q["PSI_Value_Previous_Month"]
    ) - 1
) * 100


# ------------------------------------------------------------
# 3. Total Card Payment Volume
# ------------------------------------------------------------

kpi_4q["Total_Card_Payment_Volume"] = (
    kpi_4q["CreditCard_POS_Volume"] +
    kpi_4q["CreditCard_Online_Volume"] +
    kpi_4q["DebitCard_POS_Volume"] +
    kpi_4q["DebitCard_Online_Volume"]
)


# ------------------------------------------------------------
# 4. Total QR Infrastructure
# ------------------------------------------------------------

kpi_4q["Total_QR_Infrastructure"] = (
    kpi_4q["Total_Bharat_QR"] +
    kpi_4q["Total_UPI_QR"]
)


# ------------------------------------------------------------
# 5. Total ATM Infrastructure
# ------------------------------------------------------------

kpi_4q["Total_ATM_Infrastructure"] = (
    kpi_4q["Total_ATM_Onsite"] +
    kpi_4q["Total_ATM_Offsite"]
)


# ------------------------------------------------------------
# 6. Select final KPI columns
# ------------------------------------------------------------

kpi_columns = [
    "Analysis_Period",
    "Quarter",
    "NPCI_Volume_Mn",
    "NPCI_Value_Bn",
    "NPCI_Avg_Transaction_Value",
    "PSI_Volume_Current_Month",
    "PSI_Volume_MoM_Change_%",
    "PSI_Value_Current_Month",
    "PSI_Value_MoM_Change_%",
    "Total_Card_Payment_Volume",
    "Total_QR_Infrastructure",
    "Total_ATM_Infrastructure"
]

derived_kpi_4q = kpi_4q[kpi_columns].copy()


# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print("\nDERIVED KPI TABLE")
print("-" * 90)

display(derived_kpi_4q)

print("\n" + "=" * 90)
print("4-QUARTER DERIVED KPI ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# 4-QUARTER CROSS-DATASET CORRELATION ANALYSIS
# ============================================================

print("=" * 90)
print("4-QUARTER CROSS-DATASET CORRELATION ANALYSIS")
print("=" * 90)

# Check that combined_4q exists
print("\nCombined dataset shape:", combined_4q.shape)
print("Columns:")
for i, col in enumerate(combined_4q.columns, 1):
    print(f"{i:02d}. {col}")

# Select numerical columns
correlation_columns = [
    'Total_ATM_Onsite',
    'Total_ATM_Offsite',
    'Total_POS',
    'Total_Micro_ATMs',
    'Total_Bharat_QR',
    'Total_UPI_QR',
    'Total_Credit_Cards',
    'Total_Debit_Cards',
    'CreditCard_POS_Volume',
    'CreditCard_Online_Volume',
    'DebitCard_POS_Volume',
    'DebitCard_Online_Volume',
    'DebitCard_ATM_Withdrawal_Volume',
    'PSI_Volume_Current_Month',
    'PSI_Value_Current_Month',
    'NPCI_Volume_Mn',
    'NPCI_Value_Bn'
]

# Keep only columns that actually exist
available_columns = [
    col for col in correlation_columns
    if col in combined_4q.columns
]

missing_columns = [
    col for col in correlation_columns
    if col not in combined_4q.columns
]

if missing_columns:
    print("\nMissing columns:")
    print(missing_columns)

# Correlation matrix
corr_matrix_4q = combined_4q[available_columns].corr()

print("\n" + "=" * 90)
print("4-QUARTER CORRELATION MATRIX")
print("-" * 90)
print(corr_matrix_4q.round(3))

# ------------------------------------------------------------
# Correlation with NPCI Volume
# ------------------------------------------------------------

npci_volume_corr_4q = (
    corr_matrix_4q['NPCI_Volume_Mn']
    .drop('NPCI_Volume_Mn')
    .sort_values(ascending=False)
)

print("\n" + "=" * 90)
print("CORRELATION WITH NPCI TRANSACTION VOLUME")
print("-" * 90)
print(npci_volume_corr_4q.round(3))

# ------------------------------------------------------------
# Correlation with NPCI Value
# ------------------------------------------------------------

npci_value_corr_4q = (
    corr_matrix_4q['NPCI_Value_Bn']
    .drop('NPCI_Value_Bn')
    .sort_values(ascending=False)
)

print("\n" + "=" * 90)
print("CORRELATION WITH NPCI TRANSACTION VALUE")
print("-" * 90)
print(npci_value_corr_4q.round(3))

print("\n" + "=" * 90)
print("4-QUARTER CORRELATION ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# 4-QUARTER STATISTICAL SIGNIFICANCE ANALYSIS
# ============================================================

from scipy.stats import pearsonr
import pandas as pd
import numpy as np

print("=" * 90)
print("4-QUARTER STATISTICAL SIGNIFICANCE ANALYSIS")
print("=" * 90)

target_columns = [
    'NPCI_Volume_Mn',
    'NPCI_Value_Bn'
]

feature_columns = [
    'Total_ATM_Onsite',
    'Total_ATM_Offsite',
    'Total_POS',
    'Total_Micro_ATMs',
    'Total_Bharat_QR',
    'Total_UPI_QR',
    'Total_Credit_Cards',
    'Total_Debit_Cards',
    'CreditCard_POS_Volume',
    'CreditCard_Online_Volume',
    'DebitCard_POS_Volume',
    'DebitCard_Online_Volume',
    'DebitCard_ATM_Withdrawal_Volume',
    'PSI_Volume_Current_Month',
    'PSI_Value_Current_Month'
]

results = []

for target in target_columns:

    for feature in feature_columns:

        x = combined_4q[feature].astype(float)
        y = combined_4q[target].astype(float)

        r, p = pearsonr(x, y)

        results.append({
            'Target': target,
            'Feature': feature,
            'Pearson_Correlation': r,
            'P_Value': p,
            'Significant_At_5pct': p < 0.05
        })

significance_results = pd.DataFrame(results)

significance_results = significance_results.sort_values(
    ['Target', 'Pearson_Correlation'],
    ascending=[True, False]
)

print("\nSTATISTICAL SIGNIFICANCE RESULTS")
print("-" * 90)

print(significance_results.round(4).to_string(index=False))

print("\n" + "=" * 90)
print("IMPORTANT INTERPRETATION")
print("-" * 90)
print("Number of observations:", len(combined_4q))
print("Significance level: 5%")
print("""
Because only four quarterly observations are available,
correlation and p-value estimates should be treated as exploratory.
They should NOT be interpreted as proof of causation.
""")

print("=" * 90)
print("STATISTICAL SIGNIFICANCE ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# 4-QUARTER KEY FINDINGS & BUSINESS INTERPRETATION
# ============================================================

print("=" * 90)
print("4-QUARTER KEY FINDINGS & BUSINESS INTERPRETATION")
print("=" * 90)

# ------------------------------------------------------------
# 1. QoQ growth highlights
# ------------------------------------------------------------

print("\n1. QUARTER-OVER-QUARTER HIGHLIGHTS")
print("-" * 90)

# Get the growth dataframe
growth_numeric = growth_4q.select_dtypes(include=np.number)

for col in growth_numeric.columns:
    values = growth_4q[col].dropna()

    if len(values) > 0:
        max_growth = values.max()
        min_growth = values.min()

        max_q = growth_4q.loc[values.idxmax(), 'Quarter']
        min_q = growth_4q.loc[values.idxmin(), 'Quarter']

        print(
            f"{col:45s} | "
            f"Highest: {max_growth:8.2f}% ({max_q}) | "
            f"Lowest: {min_growth:8.2f}% ({min_q})"
        )

# ------------------------------------------------------------
# 2. Overall Q1 to Q4 change
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("2. OVERALL Q1 → Q4 CHANGE")
print("-" * 90)

q1 = combined_4q.iloc[0]
q4 = combined_4q.iloc[-1]

overall_changes = []

for col in correlation_columns:
    if col in combined_4q.columns:

        start = float(q1[col])
        end = float(q4[col])

        change_pct = ((end - start) / start) * 100

        overall_changes.append({
            'Metric': col,
            'Q1_Value': start,
            'Q4_Value': end,
            'Q1_to_Q4_Change_%': change_pct
        })

overall_change_df = pd.DataFrame(overall_changes)

print(
    overall_change_df
    .sort_values('Q1_to_Q4_Change_%', ascending=False)
    .round(2)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 3. NPCI-related strongest correlations
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("3. STRONGEST RELATIONSHIPS WITH NPCI")
print("-" * 90)

print("\nNPCI Volume:")
print(npci_volume_corr_4q.abs()
      .sort_values(ascending=False)
      .head(5)
      .to_string())

print("\nNPCI Value:")
print(npci_value_corr_4q.abs()
      .sort_values(ascending=False)
      .head(5)
      .to_string())

# ------------------------------------------------------------
# 4. Significant relationships
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("4. STATISTICALLY SIGNIFICANT RELATIONSHIPS")
print("-" * 90)

significant_results = significance_results[
    significance_results['Significant_At_5pct'] == True
].copy()

print(
    significant_results[
        [
            'Target',
            'Feature',
            'Pearson_Correlation',
            'P_Value'
        ]
    ]
    .round(4)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 5. Final analytical summary
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("5. ANALYTICAL SUMMARY")
print("-" * 90)

print("""
• NPCI transaction volume increased across the four quarters.
• NPCI transaction value also increased, with the strongest increase
  occurring in Q4.
• PSI transaction volume generally increased from Q1 to Q4.
• PSI transaction value showed an overall upward trend.
• UPI QR infrastructure increased substantially across the year.
• Credit-card infrastructure and card-payment activity increased.
• Some ATM and debit-card indicators declined while digital-payment
  indicators increased.
• Strong correlations were observed between several digital-payment
  indicators and NPCI activity.
• Because the analysis contains only four quarterly observations,
  correlations and significance results are exploratory.
• Correlation does not establish causation.
""")

print("=" * 90)
print("4-QUARTER KEY FINDINGS ANALYSIS COMPLETED")
print("=" * 90)
```
