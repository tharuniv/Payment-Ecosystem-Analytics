```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
# Read the Excel file exactly as it is
raw_df = pd.read_excel("ATM_MASTER_2025.xlsx", header=None)

print("ATM Excel file loaded successfully!")
print("Number of rows:", raw_df.shape[0])
print("Number of columns:", raw_df.shape[1])
```

```python
# Display rows 0 to 20 without hiding any columns
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 30)

display(raw_df.iloc[0:20, :])
```

```python
df = raw_df.iloc[10:].copy()

df.columns = raw_df.iloc[2].values

df = df.reset_index(drop=True)

print("ATM data table created successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
# Show the header information from rows 2 to 7
header_rows = raw_df.iloc[2:8, :]

# Display all 33 columns
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 40)

display(header_rows.T)
```

```python
# Create the correct 33 column names
# based on the original multi-level headers in the ATM Excel file

column_names = [
    "Year",
    "Month",
    "Quarter",
    "Source File",
    "Sr. No.",
    "Bank Name",
    "Bank Category",

    "Infrastructure - ATMs & CRMs - On-site",
    "Infrastructure - ATMs & CRMs - Off-site",
    "Infrastructure - PoS",
    "Infrastructure - Micro ATMs",
    "Infrastructure - Bharat QR Codes",
    "Infrastructure - UPI QR Codes",
    "Infrastructure - Credit Cards",
    "Infrastructure - Debit Cards",

    "Credit Card - Card Payments - PoS - Volume (in actuals)",
    "Credit Card - Card Payments - PoS - Value (in Rs'000)",
    "Credit Card - Card Payments - Online (e-com) - Volume (in actuals)",
    "Credit Card - Card Payments - Online (e-com) - Value (in Rs'000)",
    "Credit Card - Card Payments - Others - Volume (in actuals)",
    "Credit Card - Card Payments - Others - Value (in Rs'000)",
    "Credit Card - Cash Withdrawal - ATM - Volume (in actuals)",
    "Credit Card - Cash Withdrawal - ATM - Value (in Rs'000)",

    "Debit Card - Card Payments - PoS - Volume (in actuals)",
    "Debit Card - Card Payments - PoS - Value (in Rs'000)",
    "Debit Card - Card Payments - Online (e-com) - Volume (in actuals)",
    "Debit Card - Card Payments - Online (e-com) - Value (in Rs'000)",
    "Debit Card - Card Payments - Others - Volume (in actuals)",
    "Debit Card - Card Payments - Others - Value (in Rs'000)",
    "Debit Card - Cash Withdrawal - ATM - Volume (in actuals)",
    "Debit Card - Cash Withdrawal - ATM - Value (in Rs'000)",
    "Debit Card - Cash Withdrawal - PoS - Volume (in actuals)",
    "Debit Card - Cash Withdrawal - PoS - Value (in Rs'000)"
]

# Safety check
print("Number of columns in dataset:", df.shape[1])
print("Number of new column names:", len(column_names))

if df.shape[1] == len(column_names):
    df.columns = column_names
    print("\nColumn names assigned successfully!")
else:
    print("\nERROR: Column count does not match.")
```

```python
# Check that all 33 columns are present and unique

print("Total columns:", len(df.columns))
print("Unique columns:", df.columns.nunique())

print("\nAll column names:")
for i, column in enumerate(df.columns, start=1):
    print(i, ":", column)
```

```python
# Check missing values in every column

missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": missing_values.values,
    "Missing Percentage": (missing_values.values / len(df) * 100).round(2)
})

display(missing_report)
```

```python
# Find rows where Year is missing
# We are NOT deleting anything yet

missing_year_rows = df[df["Year"].isna()]

print("Number of rows where Year is missing:", len(missing_year_rows))

display(missing_year_rows)
```

```python
# Check whether these missing-Year rows contain actual bank names

check_rows = df[
    df["Year"].isna() &
    df["Bank Name"].notna()
]

print("Rows with missing Year but Bank Name present:", len(check_rows))

display(check_rows)
```

```python
# Identify rows that are category/section rows
# These rows have a Bank Name but no Year

category_rows = df[
    df["Year"].isna() &
    df["Bank Name"].notna()
].copy()

print("Number of category rows:", len(category_rows))

print("\nCategory names found:")
print(category_rows["Bank Name"].value_counts())
```

```python
# Identify actual bank-level records
# Actual records have a Year value

bank_records = df[df["Year"].notna()].copy()

print("Total rows in dataset:", len(df))
print("Category/section rows:", len(category_rows))
print("Actual bank records:", len(bank_records))
```

```python
# Count the number of months in which each bank appears

bank_month_counts = (
    bank_records
    .groupby("Bank Name")["Month"]
    .nunique()
    .sort_values()
)

display(bank_month_counts)
```

```python
# Identify the 12 official monthly Total rows

monthly_total_records = bank_records[
    bank_records["Bank Name"].astype(str).str.strip().eq("Total")
].copy()

print("Number of monthly Total rows:", len(monthly_total_records))

display(
    monthly_total_records[
        ["Year", "Month", "Quarter", "Source File",
         "Sr. No.", "Bank Name", "Bank Category"]
    ]
)
```

```python
# Create a dataset containing only individual bank records
# Total rows are excluded from this analytical dataset

bank_data = bank_records[
    ~bank_records["Bank Name"].astype(str).str.strip().eq("Total")
].copy()

print("Actual bank-level records:", len(bank_data))
print("Columns:", len(bank_data.columns))

display(bank_data.head())
```

```python
# Columns that should contain numerical data
numeric_columns = list(bank_data.columns[7:])

# Check for non-numeric values in each numerical column
for column in numeric_columns:
    
    # Try converting to numeric without changing the original data
    converted = pd.to_numeric(bank_data[column], errors="coerce")
    
    # Find values that could not be converted
    non_numeric = bank_data.loc[
        converted.isna() & bank_data[column].notna(),
        column
    ].unique()
    
    if len(non_numeric) > 0:
        print("\nCOLUMN:", column)
        print("Non-numeric values found:", non_numeric[:20])
```

```python
# Check missing values in the 768 bank records

missing_summary = pd.DataFrame({
    "Missing Values": bank_data.isna().sum(),
    "Missing Percentage": (bank_data.isna().sum() / len(bank_data) * 100).round(2)
})

display(missing_summary)
```

```python
# Create a copy for the cleaned dataset

cleaned_atm = bank_data.copy()

# Columns that should be numeric
numeric_columns = cleaned_atm.columns[7:]

# Convert all numerical columns to numeric
for column in numeric_columns:
    cleaned_atm[column] = pd.to_numeric(
        cleaned_atm[column],
        errors="raise"
    )

# Convert Year and Sr. No. to numeric
cleaned_atm["Year"] = pd.to_numeric(
    cleaned_atm["Year"],
    errors="raise"
)

cleaned_atm["Sr. No."] = pd.to_numeric(
    cleaned_atm["Sr. No."],
    errors="raise"
)

print("Numeric conversion completed successfully.")
```

```python
# Check the data types after conversion

print(cleaned_atm.dtypes)
```

```python
# Check missing values after numeric conversion

final_missing = cleaned_atm.isna().sum()

print("Total missing values:", final_missing.sum())

if final_missing.sum() == 0:
    print("✓ No missing values were introduced.")
else:
    print("⚠ Missing values found:")
    display(final_missing[final_missing > 0])
```

```python
# Extract the 12 official monthly Total rows from the original dataset

total_rows = bank_records[
    bank_records["Bank Name"].astype(str).str.strip().eq("Total")
].copy()

print("Number of Total rows:", len(total_rows))

display(
    total_rows[["Year", "Month", "Quarter", "Bank Name"]]
)
```

```python
# Numerical columns used for total validation

numeric_columns = cleaned_atm.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("Number of numerical columns:", len(numeric_columns))

for i, col in enumerate(numeric_columns, 1):
    print(i, ":", col)
```

```python
# Calculate totals from the 64 bank records for each month

calculated_monthly_totals = (
    cleaned_atm
    .groupby("Month")[numeric_columns]
    .sum()
)

display(calculated_monthly_totals)
```

```python
# Convert official Total row numerical columns to numeric

for col in numeric_columns:
    total_rows[col] = pd.to_numeric(
        total_rows[col],
        errors="coerce"
    )

official_monthly_totals = (
    total_rows
    .set_index("Month")[numeric_columns]
)

display(official_monthly_totals)
```

```python
# Compare calculated totals with official Total rows

comparison = (
    calculated_monthly_totals
    .subtract(official_monthly_totals)
)

display(comparison)
```

```python
# Check whether calculated and official totals match
# within a very small numerical tolerance

import numpy as np

matches = np.isclose(
    calculated_monthly_totals,
    official_monthly_totals,
    rtol=1e-9,
    atol=1e-6
)

print("All totals match:", matches.all())

print(
    "Number of mismatched cells:",
    (~matches).sum().sum()
)
```

```python
# Count mismatches for each numerical column

mismatch_count = (~matches).sum(axis=0)

# Convert the NumPy array into a Pandas Series
mismatch_summary = pd.Series(
    mismatch_count,
    index=numeric_columns
)

# Keep only columns having mismatches
mismatch_summary = (
    mismatch_summary[mismatch_summary > 0]
    .sort_values(ascending=False)
)

print("Columns with mismatches:")
display(mismatch_summary)
```

```python
# Count mismatches for each month

monthly_mismatch_count = (~matches).sum(axis=1)

print("Mismatches by month:")

display(
    monthly_mismatch_count[
        monthly_mismatch_count > 0
    ]
)
```

```python
# Show the differences between our calculated totals
# and the official Total rows

difference = (
    calculated_monthly_totals
    - official_monthly_totals
)

# Show only non-zero differences
difference_nonzero = difference.where(
    ~np.isclose(
        difference,
        0,
        rtol=1e-9,
        atol=1e-6
    ),
    0
)

display(difference_nonzero)
```

```python
# Compare January calculated totals vs official totals

january_comparison = pd.DataFrame({
    "Calculated from Banks": calculated_monthly_totals.loc["January"],
    "Official Total Row": official_monthly_totals.loc["January"]
})

january_comparison["Difference"] = (
    january_comparison["Calculated from Banks"]
    - january_comparison["Official Total Row"]
)

display(january_comparison)
```
