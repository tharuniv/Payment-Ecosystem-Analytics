```python
# ============================================================
# DATASET 1 - NPCI EDA - DATA CLEANING & STANDARDIZATION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 90)
print("DATASET 1 - NPCI EDA - DATA CLEANING & STANDARDIZATION")
print("=" * 90)

# ------------------------------------------------------------
# 1. CREATE A WORKING COPY
# ------------------------------------------------------------
npci_clean = npc_df.copy()

print("\nINITIAL SHAPE")
print("-" * 90)
print(f"Rows    : {npci_clean.shape[0]}")
print(f"Columns : {npci_clean.shape[1]}")


# ------------------------------------------------------------
# 2. IDENTIFY EMBEDDED HEADER ROWS
# ------------------------------------------------------------
print("\nIDENTIFYING EMBEDDED HEADER / NON-DATA ROWS")
print("-" * 90)

header_rows = npci_clean[
    npci_clean["NPCI_Operated_System"]
    .astype(str)
    .str.strip()
    .isin(["Financial Txns:", "nan"])
]

print(f"Potential structural rows identified : {len(header_rows)}")

if len(header_rows) > 0:
    print("\nRows identified:")
    print(header_rows[["Sr_No", "NPCI_Operated_System"]])


# ------------------------------------------------------------
# 3. REMOVE ONLY THE EMBEDDED HEADER ROW
# ------------------------------------------------------------
# "Financial Txns:" is clearly a section/header row.
# Rows with missing NPCI_Operated_System are handled separately.
# We do NOT automatically remove all NaN identifier rows.

npci_clean = npci_clean[
    npci_clean["NPCI_Operated_System"]
    .astype(str)
    .str.strip()
    .ne("Financial Txns:")
].copy()

print("\nAFTER REMOVING EMBEDDED HEADER")
print("-" * 90)
print(f"Rows remaining : {len(npci_clean)}")


# ------------------------------------------------------------
# 4. STANDARDIZE COLUMN NAMES
# ------------------------------------------------------------
npci_clean.columns = (
    npci_clean.columns
    .astype(str)
    .str.strip()
)

print("\nCOLUMN STANDARDIZATION")
print("-" * 90)
print("Column names cleaned successfully.")


# ------------------------------------------------------------
# 5. IDENTIFY TRANSACTION COLUMNS
# ------------------------------------------------------------
transaction_columns = [
    col for col in npci_clean.columns
    if "_Volume_in_Mn" in col or "_Value_in_Bn" in col
]

print("\nTRANSACTION COLUMNS")
print("-" * 90)
print(f"Transaction columns : {len(transaction_columns)}")


# ------------------------------------------------------------
# 6. CONVERT TRANSACTION COLUMNS TO NUMERIC
# ------------------------------------------------------------
print("\nNUMERIC STANDARDIZATION")
print("-" * 90)

for col in transaction_columns:
    npci_clean[col] = pd.to_numeric(
        npci_clean[col],
        errors="coerce"
    )

print("All transaction columns converted to numeric format.")


# ------------------------------------------------------------
# 7. STANDARDIZE NPCI SYSTEM NAMES
# ------------------------------------------------------------
if "NPCI_Operated_System" in npci_clean.columns:
    npci_clean["NPCI_Operated_System"] = (
        npci_clean["NPCI_Operated_System"]
        .astype("string")
        .str.strip()
    )

print("\nNPCI SYSTEM NAME STANDARDIZATION")
print("-" * 90)
print("Whitespace standardized.")


# ------------------------------------------------------------
# 8. MISSING VALUES AFTER CONVERSION
# ------------------------------------------------------------
missing_after_cleaning = npci_clean.isna().sum()

print("\nMISSING VALUES AFTER CLEANING")
print("-" * 90)

missing_after_cleaning = (
    missing_after_cleaning[
        missing_after_cleaning > 0
    ]
    .sort_values(ascending=False)
)

if len(missing_after_cleaning) == 0:
    print("No missing values detected.")
else:
    print(missing_after_cleaning)


# ------------------------------------------------------------
# 9. FINAL DATA TYPES
# ------------------------------------------------------------
print("\nFINAL DATA TYPES")
print("-" * 90)

print(npci_clean.dtypes)


# ------------------------------------------------------------
# 10. FINAL CLEANED DATASET SHAPE
# ------------------------------------------------------------
print("\nFINAL CLEANED DATASET")
print("-" * 90)
print(f"Rows    : {npci_clean.shape[0]}")
print(f"Columns : {npci_clean.shape[1]}")

print("\n" + "=" * 90)
print("✓ NPCI DATA CLEANING & STANDARDIZATION COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# DATASET 1 - NPCI EDA - MISSING VALUE PATTERN ANALYSIS
# ============================================================

print("=" * 90)
print("DATASET 1 - NPCI EDA - MISSING VALUE PATTERN ANALYSIS")
print("=" * 90)

# ------------------------------------------------------------
# 1. MISSING VALUES BY ROW
# ------------------------------------------------------------
row_missing = npci_clean.isna().sum(axis=1)

print("\nMISSING VALUES BY NPCI SYSTEM")
print("-" * 90)

row_missing_summary = pd.DataFrame({
    "NPCI_Operated_System": npci_clean["NPCI_Operated_System"],
    "Missing Cells": row_missing
})

print(
    row_missing_summary
    .sort_values("Missing Cells", ascending=False)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 2. MISSING VALUES BY COLUMN TYPE
# ------------------------------------------------------------
volume_cols = [
    col for col in transaction_columns
    if "_Volume_in_Mn" in col
]

value_cols = [
    col for col in transaction_columns
    if "_Value_in_Bn" in col
]

volume_missing = npci_clean[volume_cols].isna().sum().sum()
value_missing = npci_clean[value_cols].isna().sum().sum()

print("\nMISSING VALUES BY MEASURE")
print("-" * 90)

print(f"Volume columns : {len(volume_cols)}")
print(f"Missing volume cells : {volume_missing}")

print(f"\nValue columns : {len(value_cols)}")
print(f"Missing value cells : {value_missing}")


# ------------------------------------------------------------
# 3. MISSING PERCENTAGE BY MEASURE
# ------------------------------------------------------------
total_volume_cells = npci_clean[volume_cols].size
total_value_cells = npci_clean[value_cols].size

volume_missing_pct = (
    volume_missing / total_volume_cells * 100
)

value_missing_pct = (
    value_missing / total_value_cells * 100
)

print("\nMISSING PERCENTAGE BY MEASURE")
print("-" * 90)

print(f"Volume missing percentage : {volume_missing_pct:.2f}%")
print(f"Value missing percentage  : {value_missing_pct:.2f}%")


# ------------------------------------------------------------
# 4. MISSING VALUES BY COLUMN
# ------------------------------------------------------------
column_missing = (
    npci_clean[transaction_columns]
    .isna()
    .sum()
    .sort_values(ascending=False)
)

print("\nTOP COLUMNS WITH MISSING VALUES")
print("-" * 90)

print(column_missing.to_string())


# ------------------------------------------------------------
# 5. SYSTEMS WITH MISSING VALUES IN VALUE DATA
# ------------------------------------------------------------
value_missing_by_system = npci_clean[
    ["NPCI_Operated_System"] + value_cols
].copy()

value_missing_by_system["Missing Value Cells"] = (
    value_missing_by_system[value_cols]
    .isna()
    .sum(axis=1)
)

print("\nSYSTEMS WITH HIGHEST MISSING VALUE COUNTS")
print("-" * 90)

print(
    value_missing_by_system[
        ["NPCI_Operated_System", "Missing Value Cells"]
    ]
    .sort_values("Missing Value Cells", ascending=False)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 6. CHECK COMPLETELY EMPTY TRANSACTION COLUMNS
# ------------------------------------------------------------
completely_empty_columns = [
    col for col in transaction_columns
    if npci_clean[col].isna().all()
]

print("\nCOMPLETELY EMPTY TRANSACTION COLUMNS")
print("-" * 90)

if completely_empty_columns:
    for col in completely_empty_columns:
        print(col)
else:
    print("None")


# ------------------------------------------------------------
# 7. CHECK COMPLETELY EMPTY TRANSACTION ROWS
# ------------------------------------------------------------
completely_empty_rows = npci_clean[
    npci_clean[transaction_columns].isna().all(axis=1)
]

print("\nROWS WITH NO TRANSACTION DATA")
print("-" * 90)

if len(completely_empty_rows) > 0:
    print(
        completely_empty_rows[
            ["Sr_No", "NPCI_Operated_System"]
        ].to_string(index=False)
    )
else:
    print("No completely empty transaction rows.")


print("\n" + "=" * 90)
print("✓ NPCI MISSING VALUE PATTERN ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ==========================================================================================
# DATASET 1 - NPCI EDA - DISTRIBUTION & SKEWNESS ANALYSIS
# ==========================================================================================

print("=" * 90)
print("DATASET 1 - NPCI EDA - DISTRIBUTION & SKEWNESS ANALYSIS")
print("=" * 90)

# ---------------------------------------------------------
# 1. CALCULATE SKEWNESS
# ---------------------------------------------------------

skewness = npc_df[transaction_cols].skew()

distribution_summary = pd.DataFrame({
    "Skewness": skewness,
    "Distribution": skewness.apply(
        lambda x:
        "Approximately Symmetric" if abs(x) < 0.5
        else "Moderately Right-Skewed" if x < 1
        else "Highly Right-Skewed" if x > 1
        else "Left-Skewed"
    )
})

# ---------------------------------------------------------
# 2. DISPLAY MOST SKEWED VARIABLES
# ---------------------------------------------------------

print("\nMOST SKEWED TRANSACTION VARIABLES")
print("-" * 90)

display(
    distribution_summary
    .sort_values("Skewness", ascending=False)
    .round(3)
)

# ---------------------------------------------------------
# 3. OVERALL SUMMARY
# ---------------------------------------------------------

highly_right_skewed = (skewness > 1).sum()
moderately_right_skewed = ((skewness >= 0.5) & (skewness <= 1)).sum()
approximately_symmetric = (skewness.abs() < 0.5).sum()
left_skewed = (skewness < -0.5).sum()

print("\nDISTRIBUTION SUMMARY")
print("-" * 90)

print(f"Highly right-skewed variables       : {highly_right_skewed}")
print(f"Moderately right-skewed variables   : {moderately_right_skewed}")
print(f"Approximately symmetric variables   : {approximately_symmetric}")
print(f"Left-skewed variables               : {left_skewed}")

print("\nINTERPRETATION")
print("-" * 90)

if highly_right_skewed > 0:
    print("• Several transaction variables are highly right-skewed.")
    print("• This indicates that a small number of NPCI systems")
    print("  account for substantially larger transaction activity.")

print("• Skewness will be considered during feature engineering/modeling.")
print("• No values are modified in this EDA step.")

print("\n" + "=" * 90)
print("✓ DISTRIBUTION & SKEWNESS ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ==========================================================================================
# DATASET 1 - NPCI EDA - CORRELATION ANALYSIS
# ==========================================================================================

print("=" * 90)
print("DATASET 1 - NPCI EDA - CORRELATION ANALYSIS")
print("=" * 90)

# ---------------------------------------------------------
# 1. CALCULATE CORRELATION MATRIX
# ---------------------------------------------------------

correlation_matrix = npc_df[transaction_cols].corr()

# ---------------------------------------------------------
# 2. REMOVE SELF-CORRELATIONS
# ---------------------------------------------------------

corr_pairs = correlation_matrix.where(
    ~__import__("numpy").eye(len(correlation_matrix), dtype=bool)
).stack()

# ---------------------------------------------------------
# 3. STRONGEST POSITIVE CORRELATIONS
# ---------------------------------------------------------

strongest_positive = corr_pairs.sort_values(
    ascending=False
).head(10)

print("\nTOP 10 STRONGEST POSITIVE CORRELATIONS")
print("-" * 90)

for (col1, col2), value in strongest_positive.items():
    print(f"{col1}")
    print(f"  ↳ {col2}")
    print(f"     Correlation : {value:.3f}")
    print()

# ---------------------------------------------------------
# 4. LOWEST CORRELATIONS
# ---------------------------------------------------------

lowest_correlations = corr_pairs.sort_values(
    ascending=True
).head(10)

print("\n10 LOWEST CORRELATIONS")
print("-" * 90)

for (col1, col2), value in lowest_correlations.items():
    print(f"{col1}")
    print(f"  ↳ {col2}")
    print(f"     Correlation : {value:.3f}")
    print()

# ---------------------------------------------------------
# 5. OVERALL CORRELATION RANGE
# ---------------------------------------------------------

print("\nCORRELATION SUMMARY")
print("-" * 90)

print(f"Highest correlation : {corr_pairs.max():.3f}")
print(f"Lowest correlation  : {corr_pairs.min():.3f}")
print(f"Mean correlation    : {corr_pairs.mean():.3f}")

# ---------------------------------------------------------
# 6. STRONG CORRELATION COUNT
# ---------------------------------------------------------

strong_corr_count = (corr_pairs.abs() >= 0.80).sum()

print(f"Strong correlations (|r| >= 0.80) : {strong_corr_count}")

# ---------------------------------------------------------
# 7. INTERPRETATION
# ---------------------------------------------------------

print("\nINTERPRETATION")
print("-" * 90)

print("• Correlation analysis helps identify relationships among")
print("  NPCI transaction volume and value variables.")

print("• Strong correlations may indicate that multiple transaction")
print("  measures move together over time.")

print("• Highly correlated features may contain redundant information.")
print("• Feature selection will be considered during ML preparation.")
print("• No features are removed at this stage.")

print("\n" + "=" * 90)
print("✓ CORRELATION ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ==========================================================================================
# DATASET 1 - NPCI EDA - TIME / TREND ANALYSIS
# ==========================================================================================

print("=" * 90)
print("DATASET 1 - NPCI EDA - TIME / TREND ANALYSIS")
print("=" * 90)

# ---------------------------------------------------------
# DEFINE PERIOD COLUMNS
# ---------------------------------------------------------

period_columns = [
    col for col in transaction_cols
    if "_Volume_in_Mn" in col
]

# ---------------------------------------------------------
# CALCULATE TOTAL VOLUME FOR EACH PERIOD
# ---------------------------------------------------------

period_volume = npc_df[period_columns].sum()

trend_df = pd.DataFrame({
    "Period": period_volume.index,
    "Total_Volume_Mn": period_volume.values
})

# ---------------------------------------------------------
# DISPLAY TREND
# ---------------------------------------------------------

print("\nTRANSACTION VOLUME TREND")
print("-" * 90)

display(trend_df.round(2))

# ---------------------------------------------------------
# HIGHEST AND LOWEST PERIOD
# ---------------------------------------------------------

highest_period = trend_df.loc[
    trend_df["Total_Volume_Mn"].idxmax()
]

lowest_period = trend_df.loc[
    trend_df["Total_Volume_Mn"].idxmin()
]

print("\nTREND SUMMARY")
print("-" * 90)

print(
    f"Highest volume period : "
    f"{highest_period['Period']} | "
    f"{highest_period['Total_Volume_Mn']:,.2f} Mn"
)

print(
    f"Lowest volume period  : "
    f"{lowest_period['Period']} | "
    f"{lowest_period['Total_Volume_Mn']:,.2f} Mn"
)

# ---------------------------------------------------------
# FIRST VS LAST PERIOD
# ---------------------------------------------------------

first_volume = trend_df["Total_Volume_Mn"].iloc[0]
last_volume = trend_df["Total_Volume_Mn"].iloc[-1]

if first_volume != 0:
    percentage_change = ((last_volume - first_volume) / first_volume) * 100
else:
    percentage_change = float("nan")

print(
    f"\nFirst period → Last period change : "
    f"{percentage_change:.2f}%"
)

# ---------------------------------------------------------
# INTERPRETATION
# ---------------------------------------------------------

print("\nINTERPRETATION")
print("-" * 90)

print("• Transaction volume is compared across the available")
print("  NPCI reporting periods.")

print("• The highest and lowest activity periods are identified")
print("  to understand overall transaction trends.")

print("• This trend information will later support feature")
print("  engineering and predictive modeling.")

print("• No observations are removed or modified.")

print("\n" + "=" * 90)
print("✓ TIME / TREND ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ==========================================================================================
# DATASET 1 - NPCI EDA - OUTLIER ANALYSIS
# ==========================================================================================

print("=" * 90)
print("DATASET 1 - NPCI EDA - OUTLIER ANALYSIS")
print("=" * 90)

# ---------------------------------------------------------
# 1. Make sure transaction columns are numeric
# ---------------------------------------------------------

numeric_data = npc_df[transaction_cols].apply(
    pd.to_numeric, errors="coerce"
)

# ---------------------------------------------------------
# 2. Calculate Q1, Q3 and IQR
# ---------------------------------------------------------

Q1 = numeric_data.quantile(0.25)
Q3 = numeric_data.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# ---------------------------------------------------------
# 3. Identify outliers
# ---------------------------------------------------------

outlier_mask = (
    (numeric_data < lower_bound) |
    (numeric_data > upper_bound)
)

outlier_counts = outlier_mask.sum()

# ---------------------------------------------------------
# 4. Create summary table
# ---------------------------------------------------------

outlier_summary = pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower_Bound": lower_bound,
    "Upper_Bound": upper_bound,
    "Outlier_Count": outlier_counts
})

outlier_summary["Outlier_Percentage"] = (
    outlier_summary["Outlier_Count"]
    / numeric_data.count()
    * 100
)

# ---------------------------------------------------------
# 5. Display columns containing outliers
# ---------------------------------------------------------

print("\nOUTLIER SUMMARY")
print("-" * 90)

display(
    outlier_summary[
        [
            "Q1",
            "Q3",
            "IQR",
            "Lower_Bound",
            "Upper_Bound",
            "Outlier_Count",
            "Outlier_Percentage"
        ]
    ].round(2)
)

# ---------------------------------------------------------
# 6. Overall outlier count
# ---------------------------------------------------------

total_outliers = int(outlier_counts.sum())

print("\nOVERALL OUTLIER SUMMARY")
print("-" * 90)

print("Total transaction columns :", len(transaction_cols))
print("Total outlier observations:", total_outliers)

if total_outliers > 0:
    print("\nColumns with outliers:")
    print(
        outlier_counts[outlier_counts > 0]
        .sort_values(ascending=False)
    )
else:
    print("\nNo statistical outliers detected.")

# ---------------------------------------------------------
# 7. Interpretation
# ---------------------------------------------------------

print("\nINTERPRETATION")
print("-" * 90)

print("• Outliers were identified using the IQR method.")
print("• Outliers indicate unusually high or low transaction values.")
print("• Extreme values are expected to occur because NPCI systems")
print("  have very different transaction scales.")
print("• No outliers are removed or modified during EDA.")
print("• Outlier treatment, if required, will be considered during")
print("  preprocessing and feature engineering.")

print("\n" + "=" * 90)
print("✓ OUTLIER ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ==========================================================================================
# DATASET 1 - NPCI EDA - VOLUME vs VALUE ANALYSIS
# ==========================================================================================

print("=" * 90)
print("DATASET 1 - NPCI EDA - VOLUME vs VALUE ANALYSIS")
print("=" * 90)

# ---------------------------------------------------------
# Identify volume and value columns
# ---------------------------------------------------------

volume_cols = [
    col for col in transaction_cols
    if "Volume" in col
]

value_cols = [
    col for col in transaction_cols
    if "Value" in col
]

print("\nCOLUMN SUMMARY")
print("-" * 90)
print("Volume columns :", len(volume_cols))
print("Value columns  :", len(value_cols))

# ---------------------------------------------------------
# Match volume and value columns by reporting period
# ---------------------------------------------------------

pairs = []

for volume_col in volume_cols:

    value_col = volume_col.replace(
        "Volume_in_Mn",
        "Value_in_Bn"
    )

    if value_col in value_cols:
        pairs.append((volume_col, value_col))

# ---------------------------------------------------------
# Calculate correlation for each matching period
# ---------------------------------------------------------

relationship_results = []

for volume_col, value_col in pairs:

    temp = npc_df[[volume_col, value_col]].copy()

    temp[volume_col] = pd.to_numeric(
        temp[volume_col],
        errors="coerce"
    )

    temp[value_col] = pd.to_numeric(
        temp[value_col],
        errors="coerce"
    )

    correlation = temp[volume_col].corr(
        temp[value_col]
    )

    relationship_results.append({
        "Volume_Column": volume_col,
        "Value_Column": value_col,
        "Correlation": correlation,
        "Valid_Observations": temp.dropna().shape[0]
    })

relationship_df = pd.DataFrame(
    relationship_results
)

# ---------------------------------------------------------
# Display results
# ---------------------------------------------------------

print("\nVOLUME vs VALUE CORRELATION")
print("-" * 90)

display(
    relationship_df.round(3)
)

# ---------------------------------------------------------
# Overall relationship
# ---------------------------------------------------------

valid_correlations = relationship_df["Correlation"].dropna()

print("\nRELATIONSHIP SUMMARY")
print("-" * 90)

print(
    "Periods analyzed :",
    len(relationship_df)
)

print(
    "Average correlation :",
    round(valid_correlations.mean(), 3)
)

print(
    "Highest correlation :",
    round(valid_correlations.max(), 3)
)

print(
    "Lowest correlation  :",
    round(valid_correlations.min(), 3)
)

# ---------------------------------------------------------
# Interpretation
# ---------------------------------------------------------

print("\nINTERPRETATION")
print("-" * 90)

print("• Volume represents the number of transactions.")
print("• Value represents the monetary value of transactions.")
print("• Correlation is used to examine whether higher transaction")
print("  volumes are generally associated with higher transaction values.")
print("• A strong positive correlation indicates that volume and value")
print("  tend to increase together.")
print("• This relationship can be useful during feature engineering")
print("  and predictive modeling.")
print("• No observations are removed or modified in this EDA step.")

print("\n" + "=" * 90)
print("✓ VOLUME vs VALUE ANALYSIS COMPLETED")
print("=" * 90)
```

```python
# ============================================================
# DATASET 1 - NPCI EDA - SYSTEM-WISE ANALYSIS
# ============================================================

print("=" * 90)
print("DATASET 1 - NPCI EDA - SYSTEM-WISE TRANSACTION ANALYSIS")
print("=" * 90)

# ------------------------------------------------------------
# 1. Identify volume and value columns
# ------------------------------------------------------------

volume_cols = [
    col for col in npc_df.columns
    if "_Volume_in_Mn" in col
]

value_cols = [
    col for col in npc_df.columns
    if "_Value_in_Bn" in col
]

print("\nVOLUME COLUMNS")
print("-" * 90)
print("Total volume columns :", len(volume_cols))

print("\nVALUE COLUMNS")
print("-" * 90)
print("Total value columns :", len(value_cols))


# ------------------------------------------------------------
# 2. Calculate total volume and value for each NPCI system
# ------------------------------------------------------------

system_analysis = npc_df.copy()

system_analysis["Total_Volume_Mn"] = system_analysis[volume_cols].sum(
    axis=1, skipna=True
)

system_analysis["Total_Value_Bn"] = system_analysis[value_cols].sum(
    axis=1, skipna=True
)


# ------------------------------------------------------------
# 3. Top systems by transaction volume
# ------------------------------------------------------------

top_volume = (
    system_analysis[
        ["NPCI_Operated_System", "Total_Volume_Mn"]
    ]
    .dropna(subset=["NPCI_Operated_System"])
    .sort_values("Total_Volume_Mn", ascending=False)
    .head(10)
)

print("\nTOP 10 NPCI SYSTEMS BY TOTAL TRANSACTION VOLUME")
print("-" * 90)

display(
    top_volume.round(2)
)


# ------------------------------------------------------------
# 4. Top systems by transaction value
# ------------------------------------------------------------

top_value = (
    system_analysis[
        ["NPCI_Operated_System", "Total_Value_Bn"]
    ]
    .dropna(subset=["NPCI_Operated_System"])
    .sort_values("Total_Value_Bn", ascending=False)
    .head(10)
)

print("\nTOP 10 NPCI SYSTEMS BY TOTAL TRANSACTION VALUE")
print("-" * 90)

display(
    top_value.round(2)
)


# ------------------------------------------------------------
# 5. Overall system-wise summary
# ------------------------------------------------------------

system_summary = (
    system_analysis[
        [
            "NPCI_Operated_System",
            "Total_Volume_Mn",
            "Total_Value_Bn"
        ]
    ]
    .dropna(subset=["NPCI_Operated_System"])
    .sort_values("Total_Volume_Mn", ascending=False)
)

print("\nSYSTEM-WISE TRANSACTION SUMMARY")
print("-" * 90)

display(
    system_summary.round(2)
)


# ------------------------------------------------------------
# 6. Interpretation
# ------------------------------------------------------------

print("\nINTERPRETATION")
print("-" * 90)

print("• NPCI systems were compared based on their total transaction volume and value.")
print("• The analysis identifies the systems contributing the highest transaction activity.")
print("• High-volume systems represent major contributors to transaction frequency.")
print("• High-value systems represent major contributors to monetary transaction activity.")
print("• System-wise differences will help in feature engineering and predictive modeling.")
print("• No observations were removed or modified in this EDA step.")

print("\n" + "=" * 90)
print("✓ SYSTEM-WISE TRANSACTION ANALYSIS COMPLETED")
print("=" * 90)
```
