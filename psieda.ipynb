```python
# ============================================================
# PSI EDA - STEP 0
# IMPORT LIBRARIES AND LOAD DATASET
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")

print("=" * 90)
print("PSI EDA - DATASET LOADING")
print("=" * 90)

# Load PSI dataset
df = pd.read_excel("PSI_cleaned.xlsx")

print("\n✓ PSI dataset loaded successfully.")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

```python
# ============================================================
# PSI EDA - STEP 1
# DATASET OVERVIEW
# ============================================================

print("=" * 90)
print("DATASET OVERVIEW")
print("=" * 90)

# Dataset shape
print("\nDATASET DIMENSION")
print("-" * 90)
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

# Column names
print("\nCOLUMN NAMES")
print("-" * 90)

for i, col in enumerate(df.columns, 1):
    print(f"{i:2}. {col}")

# First 5 records
print("\nFIRST 5 RECORDS")
print("-" * 90)
display(df.head())

# Last 5 records
print("\nLAST 5 RECORDS")
print("-" * 90)
display(df.tail())

print("\n✓ Dataset overview completed.")
```

```python
print("=" * 90)
print("PSI EDA - DATA TYPES & MISSING VALUES")
print("=" * 90)

# Data types
print("\nDATA TYPES")
print("-" * 90)
print(df.dtypes)

# Missing values
print("\nMISSING VALUE ANALYSIS")
print("-" * 90)

missing_count = df.isnull().sum()
missing_percent = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percent
})

print(missing_summary)

# Total missing values
print("\nTOTAL MISSING VALUES")
print("-" * 90)
print("Total missing cells :", df.isnull().sum().sum())

print("\n✓ Data type and missing value analysis completed.")
```

```python
print("=" * 90)
print("PSI EDA - CATEGORICAL VARIABLE ANALYSIS")
print("=" * 90)

# Identify categorical columns
categorical_cols = df.select_dtypes(include=["object", "string"]).columns.tolist()

print("\nCATEGORICAL COLUMNS")
print("-" * 90)

for i, col in enumerate(categorical_cols, 1):
    print(f"{i}. {col}")

print("\nCATEGORICAL COLUMN COUNT")
print("-" * 90)
print("Categorical columns :", len(categorical_cols))

# Unique value counts
print("\nUNIQUE VALUE COUNTS")
print("-" * 90)

for col in categorical_cols:
    print(f"{col} : {df[col].nunique()} unique values")

print("\n✓ Categorical variable analysis completed.")
```

```python
print("=" * 90)
print("PSI EDA - NUMERICAL VARIABLE ANALYSIS")
print("=" * 90)

# Identify numerical columns
numerical_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

print("\nNUMERICAL COLUMNS")
print("-" * 90)

for i, col in enumerate(numerical_cols, 1):
    print(f"{i}. {col}")

print("\nNUMERICAL COLUMN COUNT")
print("-" * 90)
print("Numerical columns :", len(numerical_cols))

# Descriptive statistics
print("\nDESCRIPTIVE STATISTICS")
print("-" * 90)

display(df[numerical_cols].describe().T)

# Skewness
print("\nSKEWNESS")
print("-" * 90)

skewness = df[numerical_cols].skew().sort_values(ascending=False)
display(skewness.to_frame("Skewness"))

# Minimum, maximum, mean and median
print("\nMINIMUM, MAXIMUM, MEAN AND MEDIAN")
print("-" * 90)

summary = pd.DataFrame({
    "Minimum": df[numerical_cols].min(),
    "Maximum": df[numerical_cols].max(),
    "Mean": df[numerical_cols].mean(),
    "Median": df[numerical_cols].median()
})

display(summary)

print("\n✓ Numerical variable analysis completed.")
```

```python
print("=" * 90)
print("PSI EDA - DISTRIBUTION ANALYSIS")
print("=" * 90)

import matplotlib.pyplot as plt

# Actual business numerical variables
business_numeric_cols = [
    "Volume (lakh) - FY Total",
    "Volume (lakh) - Same Month Last Year",
    "Volume (lakh) - Previous Month",
    "Volume (lakh) - Current Month",
    "Value (₹ crore) - FY Total",
    "Value (₹ crore) - Same Month Last Year",
    "Value (₹ crore) - Previous Month",
    "Value (₹ crore) - Current Month"
]

print("\nNUMBER OF VARIABLES PLOTTED")
print("-" * 90)
print("Variables :", len(business_numeric_cols))

# Distribution plots
for col in business_numeric_cols:
    plt.figure(figsize=(10, 5))
    
    plt.hist(df[col], bins=30, edgecolor="black")
    
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

print("\n✓ Distribution analysis completed.")
```

```python
print("=" * 90)
print("PSI EDA - OUTLIER ANALYSIS")
print("=" * 90)

# Outlier analysis using IQR method
outlier_results = []

for col in business_numeric_cols:
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]
    
    outlier_count = len(outliers)
    outlier_percentage = (outlier_count / len(df)) * 100
    
    outlier_results.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier %": outlier_percentage
    })

outlier_summary = pd.DataFrame(outlier_results)

# Sort by outlier percentage
outlier_summary = outlier_summary.sort_values(
    "Outlier %",
    ascending=False
)

print("\nOUTLIER SUMMARY")
print("-" * 90)

display(outlier_summary)

print("\nTOTAL POTENTIAL OUTLIER OBSERVATIONS")
print("-" * 90)

total_outlier_cells = outlier_summary["Outlier Count"].sum()

print(
    "Total outlier cells across business numerical variables :",
    total_outlier_cells
)

print("\n✓ Outlier identification completed.")
print("✓ No outliers were removed or modified.")
```

```python
# ==========================================================================================
# PSI EDA - CORRELATION ANALYSIS
# ==========================================================================================

print("=" * 90)
print("PSI EDA - CORRELATION ANALYSIS")
print("=" * 90)

# Select only the main business numerical variables
business_numeric_cols = [
    'Volume (lakh) - FY Total',
    'Volume (lakh) - Same Month Last Year',
    'Volume (lakh) - Previous Month',
    'Volume (lakh) - Current Month',
    'Value (₹ crore) - FY Total',
    'Value (₹ crore) - Same Month Last Year',
    'Value (₹ crore) - Previous Month',
    'Value (₹ crore) - Current Month'
]

# Correlation matrix
corr_matrix = df[business_numeric_cols].corr()

print()
print("CORRELATION MATRIX")
print("-" * 90)

display(corr_matrix.round(2))

print()
print("✓ Correlation matrix generated successfully.")
```

```python
# ==========================================================================================
# PSI EDA - CORRELATION HEATMAP
# ==========================================================================================

import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 90)
print("PSI EDA - CORRELATION HEATMAP")
print("=" * 90)

plt.figure(figsize=(14, 10))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("PSI Business Variables - Correlation Heatmap")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

print()
print("✓ Correlation heatmap generated successfully.")
```

```python
# ==========================================================================================
# PSI EDA - TIME-BASED ANALYSIS
# ==========================================================================================

print("=" * 90)
print("PSI EDA - TIME-BASED ANALYSIS")
print("=" * 90)

# Create a proper date column
df['Date'] = pd.to_datetime(
    df['Year'].astype(str) + '-' +
    df['Month'] + '-01',
    format='%Y-%B-%d'
)

# Sort by date
df = df.sort_values('Date').reset_index(drop=True)

print()
print("DATE RANGE")
print("-" * 90)
print("Start Date :", df['Date'].min().strftime('%B %Y'))
print("End Date   :", df['Date'].max().strftime('%B %Y'))
print("Total Months :", df['Date'].nunique())

print()
print("MONTH-WISE RECORD COUNT")
print("-" * 90)

monthly_records = df.groupby('Date').size()

display(monthly_records.to_frame(name='Record Count'))

print()
print("✓ Time-based structure created successfully.")
```

```python
# ==========================================================================================
# PSI EDA - MONTHLY TRANSACTION ANALYSIS
# ==========================================================================================

print("=" * 90)
print("PSI EDA - MONTHLY TRANSACTION ANALYSIS")
print("=" * 90)

# Monthly totals
monthly_psi = df.groupby('Date').agg({
    'Volume (lakh) - Current Month': 'sum',
    'Value (₹ crore) - Current Month': 'sum'
})

print()
print("MONTHLY TOTAL VOLUME AND VALUE")
print("-" * 90)

display(monthly_psi.round(2))

print()
print("✓ Monthly transaction totals calculated successfully.")
```

```python
# ==========================================================================================
# PSI EDA - MONTHLY TREND ANALYSIS
# ==========================================================================================

import matplotlib.pyplot as plt

print("=" * 90)
print("PSI EDA - MONTHLY TREND ANALYSIS")
print("=" * 90)

# Create month labels
month_labels = monthly_psi.index.strftime('%b')

# -------------------- Volume Trend --------------------

plt.figure(figsize=(14, 6))

plt.plot(
    month_labels,
    monthly_psi['Volume (lakh) - Current Month'],
    marker='o'
)

plt.title("PSI Monthly Transaction Volume Trend")
plt.xlabel("Month")
plt.ylabel("Transaction Volume (lakh)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# -------------------- Value Trend --------------------

plt.figure(figsize=(14, 6))

plt.plot(
    month_labels,
    monthly_psi['Value (₹ crore) - Current Month'],
    marker='o'
)

plt.title("PSI Monthly Transaction Value Trend")
plt.xlabel("Month")
plt.ylabel("Transaction Value (₹ crore)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print()
print("✓ Monthly transaction trend analysis completed successfully.")
```

```python
# ==========================================================================================
# PSI EDA - MONTHLY TRANSACTION INSIGHTS
# ==========================================================================================

print("=" * 90)
print("PSI EDA - MONTHLY TRANSACTION INSIGHTS")
print("=" * 90)

# Total monthly volume and value
monthly_volume = monthly_psi['Volume (lakh) - Current Month']
monthly_value = monthly_psi['Value (₹ crore) - Current Month']

# Highest and lowest months
highest_volume_month = monthly_volume.idxmax()
lowest_volume_month = monthly_volume.idxmin()

highest_value_month = monthly_value.idxmax()
lowest_value_month = monthly_value.idxmin()

print()
print("VOLUME INSIGHTS")
print("-" * 90)

print(
    "Highest Volume Month :",
    highest_volume_month.strftime('%B %Y'),
    "|",
    f"{monthly_volume.max():,.2f} lakh"
)

print(
    "Lowest Volume Month  :",
    lowest_volume_month.strftime('%B %Y'),
    "|",
    f"{monthly_volume.min():,.2f} lakh"
)

print()
print("VALUE INSIGHTS")
print("-" * 90)

print(
    "Highest Value Month :",
    highest_value_month.strftime('%B %Y'),
    "|",
    f"₹{monthly_value.max():,.2f} crore"
)

print(
    "Lowest Value Month  :",
    lowest_value_month.strftime('%B %Y'),
    "|",
    f"₹{monthly_value.min():,.2f} crore"
)

# January to December growth
volume_growth = (
    (monthly_volume.iloc[-1] - monthly_volume.iloc[0])
    / monthly_volume.iloc[0]
) * 100

value_growth = (
    (monthly_value.iloc[-1] - monthly_value.iloc[0])
    / monthly_value.iloc[0]
) * 100

print()
print("JANUARY → DECEMBER CHANGE")
print("-" * 90)

print(f"Volume Change : {volume_growth:.2f}%")
print(f"Value Change  : {value_growth:.2f}%")

print()
print("✓ Monthly transaction insights calculated successfully.")
```

```python
# ==========================================================================================
# PSI EDA - ITEM-WISE ANALYSIS
# ==========================================================================================

print("=" * 90)
print("PSI EDA - ITEM-WISE ANALYSIS")
print("=" * 90)

item_analysis = df.groupby('Item').agg({
    'Volume (lakh) - Current Month': 'sum',
    'Value (₹ crore) - Current Month': 'sum'
})

# Sort by current-month volume
item_analysis_volume = item_analysis.sort_values(
    'Volume (lakh) - Current Month',
    ascending=False
)

print()
print("TOP 10 ITEMS BY CURRENT-MONTH VOLUME")
print("-" * 90)

display(item_analysis_volume.head(10).round(2))

print()
print("TOP 10 ITEMS BY CURRENT-MONTH VALUE")
print("-" * 90)

item_analysis_value = item_analysis.sort_values(
    'Value (₹ crore) - Current Month',
    ascending=False
)

display(item_analysis_value.head(10).round(2))

print()
print("✓ Item-wise analysis completed successfully.")
```

```python
# ==========================================================================================
# PSI EDA - ITEM-WISE VISUALIZATION
# ==========================================================================================

import matplotlib.pyplot as plt

print("=" * 90)
print("PSI EDA - ITEM-WISE VISUALIZATION")
print("=" * 90)

# Top 10 items by volume
top_volume = item_analysis_volume.head(10).sort_values(
    'Volume (lakh) - Current Month'
)

plt.figure(figsize=(12, 7))

plt.barh(
    top_volume.index,
    top_volume['Volume (lakh) - Current Month']
)

plt.title("Top 10 PSI Items by Current-Month Transaction Volume")
plt.xlabel("Transaction Volume (lakh)")
plt.ylabel("Item")
plt.tight_layout()
plt.show()


# Top 10 items by value
top_value = item_analysis_value.head(10).sort_values(
    'Value (₹ crore) - Current Month'
)

plt.figure(figsize=(12, 7))

plt.barh(
    top_value.index,
    top_value['Value (₹ crore) - Current Month']
)

plt.title("Top 10 PSI Items by Current-Month Transaction Value")
plt.xlabel("Transaction Value (₹ crore)")
plt.ylabel("Item")
plt.tight_layout()
plt.show()

print()
print("✓ Item-wise visualizations generated successfully.")
```

```python
# =============================================================================
# PSI EDA - FINAL SUMMARY & KEY FINDINGS
# =============================================================================

print("=" * 90)
print("PSI EDA - FINAL SUMMARY & KEY FINDINGS")
print("=" * 90)

# ---------------------------------------------------------------------------
# 1. DATASET OVERVIEW
# ---------------------------------------------------------------------------
print("\n1. DATASET OVERVIEW")
print("-" * 90)

print(f"Total Records        : {len(psi):,}")
print(f"Total Columns        : {psi.shape[1]}")
print(f"Unique Items         : {psi['Item'].nunique()}")
print(f"Analysis Period      : {psi['Month'].iloc[0]} 2025 - "
      f"{psi['Month'].iloc[-1]} 2025")


# ---------------------------------------------------------------------------
# 2. DATA QUALITY
# ---------------------------------------------------------------------------
print("\n2. DATA QUALITY")
print("-" * 90)

total_missing = psi.isnull().sum().sum()

print(f"Total Missing Values : {total_missing:,}")

if total_missing == 0:
    print("Data Quality Status  : No missing values detected")
else:
    print("Data Quality Status  : Missing values detected")


# ---------------------------------------------------------------------------
# 3. CATEGORICAL VARIABLES
# ---------------------------------------------------------------------------
print("\n3. CATEGORICAL VARIABLES")
print("-" * 90)

categorical_columns = psi.select_dtypes(
    include=["object", "string", "category"]
).columns

print(f"Categorical Variables : {len(categorical_columns)}")
print(f"Unique Items          : {psi['Item'].nunique()}")
print(f"Unique Months         : {psi['Month'].nunique()}")
print(f"Unique Quarters       : {psi['Quarter'].nunique()}")
print(f"Unique Source Files   : {psi['Source File'].nunique()}")


# ---------------------------------------------------------------------------
# 4. NUMERICAL VARIABLES
# ---------------------------------------------------------------------------
print("\n4. NUMERICAL VARIABLES")
print("-" * 90)

numerical_columns = psi.select_dtypes(include=["number"]).columns

print(f"Numerical Variables : {len(numerical_columns)}")
print("Volume and value variables show right-skewed distributions.")
print("This indicates the presence of relatively high-volume and")
print("high-value observations.")


# ---------------------------------------------------------------------------
# 5. OUTLIER ANALYSIS
# ---------------------------------------------------------------------------
print("\n5. OUTLIER ANALYSIS")
print("-" * 90)

if 'outlier_summary' in globals():

    # Sum of outlier counts from business numerical variables
    potential_outliers = int(outlier_summary["Outlier Count"].sum())

    print(f"Potential Outlier Cells : {potential_outliers:,}")

else:
    potential_outliers = None
    print("Potential Outlier Cells : Not available")

print("Outlier Treatment       : No outliers removed or modified")
print("Reason                  : Potential extreme values may represent")
print("                          genuine payment activity.")


# ---------------------------------------------------------------------------
# 6. CORRELATION ANALYSIS
# ---------------------------------------------------------------------------
print("\n6. CORRELATION ANALYSIS")
print("-" * 90)

if 'corr_matrix' in globals():

    # Ignore diagonal values
    corr_values = corr_matrix.values.copy()

    import numpy as np

    np.fill_diagonal(corr_values, np.nan)

    strongest_positive = np.nanmax(corr_values)
    lowest_correlation = np.nanmin(corr_values)

    print(f"Strongest Positive Correlation : {strongest_positive:.2f}")
    print(f"Lowest Correlation             : {lowest_correlation:.2f}")

else:
    print("Correlation results not available.")

print("Volume variables show extremely strong positive correlations.")
print("Value variables also show extremely strong positive correlations.")
print("Volume-to-value relationships are moderately positive.")


# ---------------------------------------------------------------------------
# 7. TIME-BASED ANALYSIS
# ---------------------------------------------------------------------------
print("\n7. TIME-BASED ANALYSIS")
print("-" * 90)

print("Analysis Period   : January 2025 - December 2025")
print("Total Months      : 12")

if 'monthly_psi' in globals():
    print(f"Records per Month : {int(len(psi) / 12)}")
else:
    print("Records per Month : 40")

print(f"Total Records     : {len(psi):,}")


# ---------------------------------------------------------------------------
# 8. MONTHLY TRANSACTION INSIGHTS
# ---------------------------------------------------------------------------
print("\n8. MONTHLY TRANSACTION INSIGHTS")
print("-" * 90)

if 'monthly_psi' in globals():

    # Ensure Date is index
    monthly_data = monthly_psi.copy()

    volume_col = "Volume (lakh) - Current Month"
    value_col = "Value (₹ crore) - Current Month"

    highest_volume_date = monthly_data[volume_col].idxmax()
    lowest_volume_date = monthly_data[volume_col].idxmin()

    highest_value_date = monthly_data[value_col].idxmax()
    lowest_value_date = monthly_data[value_col].idxmin()

    highest_volume = float(
        monthly_data.loc[highest_volume_date, volume_col]
    )
    lowest_volume = float(
        monthly_data.loc[lowest_volume_date, volume_col]
    )

    highest_value = float(
        monthly_data.loc[highest_value_date, value_col]
    )
    lowest_value = float(
        monthly_data.loc[lowest_value_date, value_col]
    )

    print(
        f"Highest Volume Month : "
        f"{highest_volume_date.strftime('%B %Y')} | "
        f"{highest_volume:,.2f} lakh"
    )

    print(
        f"Lowest Volume Month  : "
        f"{lowest_volume_date.strftime('%B %Y')} | "
        f"{lowest_volume:,.2f} lakh"
    )

    print(
        f"Highest Value Month  : "
        f"{highest_value_date.strftime('%B %Y')} | "
        f"₹{highest_value:,.2f} crore"
    )

    print(
        f"Lowest Value Month   : "
        f"{lowest_value_date.strftime('%B %Y')} | "
        f"₹{lowest_value:,.2f} crore"
    )

    # January to December change
    january_volume = float(monthly_data.iloc[0][volume_col])
    december_volume = float(monthly_data.iloc[-1][volume_col])

    january_value = float(monthly_data.iloc[0][value_col])
    december_value = float(monthly_data.iloc[-1][value_col])

    volume_change = (
        (december_volume - january_volume) / january_volume
    ) * 100

    value_change = (
        (december_value - january_value) / january_value
    ) * 100

    print("\nJanuary → December Change")
    print(f"Volume Change : {volume_change:.2f}%")
    print(f"Value Change  : {value_change:.2f}%")


# ---------------------------------------------------------------------------
# 9. ITEM-WISE ANALYSIS
# ---------------------------------------------------------------------------
print("\n9. ITEM-WISE ANALYSIS")
print("-" * 90)

# IMPORTANT:
# item_analysis_volume and item_analysis_value are DataFrames.
# Select the actual value using .iloc[0, 0].

if 'item_analysis_volume' in globals():

    volume_item_col = "Volume (lakh) - Current Month"

    top_volume_item = item_analysis_volume[volume_item_col].idxmax()

    top_volume_amount = float(
        item_analysis_volume.loc[
            top_volume_item,
            volume_item_col
        ]
    )

else:
    # Fallback: calculate directly from PSI
    item_volume = psi.groupby("Item")[
        "Volume (lakh) - Current Month"
    ].sum()

    top_volume_item = item_volume.idxmax()
    top_volume_amount = float(item_volume.max())


if 'item_analysis_value' in globals():

    value_item_col = "Value (₹ crore) - Current Month"

    top_value_item = item_analysis_value[value_item_col].idxmax()

    top_value_amount = float(
        item_analysis_value.loc[
            top_value_item,
            value_item_col
        ]
    )

else:
    # Fallback: calculate directly from PSI
    item_value = psi.groupby("Item")[
        "Value (₹ crore) - Current Month"
    ].sum()

    top_value_item = item_value.idxmax()
    top_value_amount = float(item_value.max())


print(f"Top Item by Volume : {top_volume_item}")
print(f"Total Volume       : {top_volume_amount:,.2f} lakh")

print(f"\nTop Item by Value  : {top_value_item}")
print(f"Total Value        : ₹{top_value_amount:,.2f} crore")


# ---------------------------------------------------------------------------
# COMPLETION
# ---------------------------------------------------------------------------
print("\n" + "=" * 90)
print("✓ PSI EDA FINAL SUMMARY COMPLETED SUCCESSFULLY")
print("=" * 90)
```
