```python
import pandas as pd
import numpy as np

file_path = r"PSI_MASTER_2025 (1).xlsx"

xls = pd.ExcelFile(file_path)

print("Sheets:")
for sheet in xls.sheet_names:
    print("-", sheet)
```

```python
sheets = {}

for sheet in xls.sheet_names:
    df = pd.read_excel(
        file_path,
        sheet_name=sheet,
        header=1
    )
    
    sheets[sheet] = df
    
    print(f"{sheet}: {df.shape}")
    print(df.columns.tolist())
    print()
```

```python
for sheet, df in sheets.items():
    print("\n" + "="*70)
    print(sheet)
    print("="*70)
    print("Shape:", df.shape)
    print("Columns:")
    
    for i, col in enumerate(df.columns):
        print(i, ":", repr(col))
```

```python
for sheet, df in sheets.items():
    print("\n" + "="*70)
    print(sheet)
    print("="*70)

    print("Total rows:", len(df))

    if "Item" in df.columns:
        print("Missing Item:", df["Item"].isna().sum())
        
        display(
            df[df["Item"].isna()].head(20)
        )

    elif sheet == "Part V - Domestic Frauds":
        print("Missing Month:", df["Month"].isna().sum())
        
        display(
            df[df["Month"].isna()].head(20)
        )
```

```python
# Step 5: Detailed missing-value check

for sheet, df in sheets.items():

    print("\n" + "="*70)
    print(sheet)
    print("="*70)

    missing = df.isna().sum()
    missing_pct = (missing / len(df) * 100).round(2)

    missing_summary = pd.DataFrame({
        "Missing Values": missing,
        "Missing %": missing_pct
    })

    missing_summary = missing_summary[
        missing_summary["Missing Values"] > 0
    ].sort_values("Missing Values", ascending=False)

    if missing_summary.empty:
        print("No missing values.")
    else:
        display(missing_summary)
```

```python
# Step 6: Remove structural/header rows

cleaned_sheets = {}

for sheet, df in sheets.items():

    df = df.copy()

    if "Year" in df.columns:
        # Keep only actual monthly observations
        df = df[df["Year"].notna()].copy()

    elif "Report Year" in df.columns:
        # Part V: keep only actual fraud observations
        df = df[df["Report Year"].notna()].copy()

    cleaned_sheets[sheet] = df.reset_index(drop=True)

    print(f"{sheet}:")
    print("  Original rows :", len(sheets[sheet]))
    print("  Cleaned rows  :", len(df))
    print("  Removed rows  :", len(sheets[sheet]) - len(df))
```

```python
for sheet, df in sheets.items():

    print("\n" + "="*70)
    print(sheet)
    print("="*70)

    print("Total rows:", len(df))
    print("Exact duplicate rows:", df.duplicated().sum())

    if df.duplicated().sum() > 0:
        print("\nDuplicate rows:")
        display(
            df[df.duplicated(keep=False)]
            .sort_values(by=df.columns.tolist())
            .head(30)
        )
```

```python
for sheet, df in sheets.items():

    print("\n" + "="*70)
    print(sheet)
    print("="*70)

    if sheet == "Part V - Domestic Frauds":
        key_cols = ["Month", "Report Year", "Report Month"]
    else:
        key_cols = ["Item", "Year", "Month"]

    duplicate_count = df.duplicated(
        subset=key_cols,
        keep=False
    ).sum()

    print("Duplicate observations based on:", key_cols)
    print("Duplicate rows:", duplicate_count)

    if duplicate_count > 0:
        display(
            df[df.duplicated(subset=key_cols, keep=False)]
            .sort_values(key_cols)
            .head(30)
        )
```

```python
# Create cleaned copies of all sheets
# We remove only rows where the time/report Year is missing.
# These are the structural/header rows identified earlier.

sheets_cleaned = {}

for sheet, df in sheets.items():

    df = df.copy()

    if sheet == "Part V - Domestic Frauds":
        # Part V uses Report Year
        df_clean = df[df["Report Year"].notna()].copy()
    else:
        # Parts I-IV use Year
        df_clean = df[df["Year"].notna()].copy()

    sheets_cleaned[sheet] = df_clean

    print(f"{sheet}:")
    print(f"  Original rows : {len(df)}")
    print(f"  Cleaned rows  : {len(df_clean)}")
    print(f"  Removed rows  : {len(df) - len(df_clean)}")
```

```python
for sheet, df in sheets_cleaned.items():

    print("\n" + "="*70)
    print(sheet)
    print("="*70)

    if sheet == "Part V - Domestic Frauds":
        print("Missing Report Year:", df["Report Year"].isna().sum())
    else:
        print("Missing Year:", df["Year"].isna().sum())
```

```python
# Step 2: Check exact duplicate rows in the cleaned data

for sheet, df in sheets_cleaned.items():

    print("\n" + "="*70)
    print(sheet)
    print("="*70)

    duplicate_count = df.duplicated().sum()

    print("Exact duplicate rows:", duplicate_count)

    if duplicate_count > 0:
        display(df[df.duplicated(keep=False)].head(20))
```

```python
# Step 3: Check data types

for sheet, df in sheets_cleaned.items():

    print("\n" + "="*70)
    print(sheet)
    print("="*70)

    display(
        pd.DataFrame({
            "Column": df.columns,
            "Data Type": df.dtypes.astype(str).values,
            "Non-Null Values": df.notna().sum().values
        })
    )
```

```python
# Step 4: Identify completely empty columns

for sheet, df in sheets_cleaned.items():

    print("\n" + "="*70)
    print(sheet)
    print("="*70)

    empty_cols = df.columns[df.isna().all()].tolist()

    print("Completely empty columns:")
    print(empty_cols)
```

```python
# Remove columns that contain no data at all

for sheet in sheets_cleaned:

    df = sheets_cleaned[sheet]

    empty_cols = df.columns[df.isna().all()].tolist()

    if empty_cols:
        sheets_cleaned[sheet] = df.drop(columns=empty_cols)

        print(f"{sheet}: removed {len(empty_cols)} empty column(s)")
        print(empty_cols)
    else:
        print(f"{sheet}: no completely empty columns")
```

```python
# Step 5: Check negative and zero values in numerical columns

for sheet, df in sheets_cleaned.items():

    print("\n" + "="*70)
    print(sheet)
    print("="*70)

    numeric_cols = df.select_dtypes(include="number").columns

    for col in numeric_cols:

        negative_count = (df[col] < 0).sum()
        zero_count = (df[col] == 0).sum()

        if negative_count > 0 or zero_count > 0:

            print(f"\n{col}")
            print("  Negative values:", negative_count)
            print("  Zero values    :", zero_count)
```

```python
df = sheets_cleaned["Part I - Payment Indicators"]

numeric_cols = [
    "Volume (lakh) - FY Total",
    "Volume (lakh) - Same Month Last Year",
    "Volume (lakh) - Previous Month",
    "Volume (lakh) - Current Month",
    "Value (₹ crore) - FY Total",
    "Value (₹ crore) - Same Month Last Year",
    "Value (₹ crore) - Previous Month",
    "Value (₹ crore) - Current Month"
]

# Find rows where all payment volume/value measures are zero
zero_rows = df[(df[numeric_cols] == 0).all(axis=1)]

print("Rows where all payment measures are zero:", len(zero_rows))

display(
    zero_rows[
        ["Item", "Year", "Month", "Quarter"] + numeric_cols
    ]
)
```

```python
# Check categorical and temporal consistency

month_order = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

quarter_order = {
    "January": "Q1",
    "February": "Q1",
    "March": "Q1",
    "April": "Q2",
    "May": "Q2",
    "June": "Q2",
    "July": "Q3",
    "August": "Q3",
    "September": "Q3",
    "October": "Q4",
    "November": "Q4",
    "December": "Q4"
}

for sheet, df in sheets_cleaned.items():

    print("\n" + "=" * 70)
    print(sheet)
    print("=" * 70)

    if sheet == "Part V - Domestic Frauds":
        year_col = "Report Year"
        month_col = "Report Month"
        quarter_col = "Report Quarter"
    else:
        year_col = "Year"
        month_col = "Month"
        quarter_col = "Quarter"

    # Check invalid months
    invalid_months = df[
        ~df[month_col].isin(month_order.keys())
    ]

    print("Invalid months:", len(invalid_months))

    if len(invalid_months) > 0:
        display(invalid_months[[month_col]].drop_duplicates())

    # Check invalid quarters
    invalid_quarters = df[
        ~df[quarter_col].isin(["Q1", "Q2", "Q3", "Q4"])
    ]

    print("Invalid quarters:", len(invalid_quarters))

    if len(invalid_quarters) > 0:
        display(invalid_quarters[[quarter_col]].drop_duplicates())

    # Check Month → Quarter consistency
    expected_quarter = df[month_col].map(quarter_order)

    inconsistent_quarter = df[
        expected_quarter.notna() &
        (df[quarter_col] != expected_quarter)
    ]

    print("Month-Quarter inconsistencies:", len(inconsistent_quarter))

    if len(inconsistent_quarter) > 0:
        display(
            inconsistent_quarter[
                [month_col, quarter_col]
            ].drop_duplicates()
        )
```

```python
# Check numeric columns for invalid / non-numeric values

for sheet, df in sheets_cleaned.items():

    print("\n" + "=" * 70)
    print(sheet)
    print("=" * 70)

    # Identify numeric-looking columns
    numeric_columns = []

    for col in df.columns:
        if any(keyword in col.lower() for keyword in [
            "volume",
            "value",
            "year",
            "fraud",
            "fts",
            "one in every"
        ]):
            numeric_columns.append(col)

    for col in numeric_columns:

        # Try converting to numeric without changing original data
        converted = pd.to_numeric(df[col], errors="coerce")

        # Values that were not numeric
        invalid = df[col].notna() & converted.isna()

        print(f"\n{col}")
        print("  Data type       :", df[col].dtype)
        print("  Non-null values :", df[col].notna().sum())
        print("  Non-numeric     :", invalid.sum())

        if invalid.sum() > 0:
            print("  Examples:")
            print(df.loc[invalid, col].unique()[:10])
```

```python
# Convert FTS from text like "0.127 bps" into a numeric column

fraud_df = sheets_cleaned["Part V - Domestic Frauds"].copy()

fraud_df["FTS (bps)"] = (
    fraud_df["FTS (Fraud / Payment Value * 10000)"]
    .str.replace("bps", "", regex=False)
    .str.strip()
    .astype(float)
)

print("Original FTS column:")
print(fraud_df["FTS (Fraud / Payment Value * 10000)"].head())

print("\nConverted FTS (bps):")
print(fraud_df["FTS (bps)"].head())

print("\nData type:")
print(fraud_df["FTS (bps)"].dtype)

print("\nMissing values:")
print(fraud_df["FTS (bps)"].isna().sum())
```

```python
sheets_cleaned["Part V - Domestic Frauds"] = fraud_df
```

```python
# Check numerical ranges and potential outliers

for sheet, df in sheets_cleaned.items():

    print("\n" + "=" * 70)
    print(sheet)
    print("=" * 70)

    numeric_cols = df.select_dtypes(include="number").columns

    for col in numeric_cols:

        s = df[col].dropna()

        if len(s) == 0:
            continue

        q1 = s.quantile(0.25)
        q3 = s.quantile(0.75)
        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        outliers = s[(s < lower) | (s > upper)]

        print(f"\n{col}")
        print(f"  Min       : {s.min()}")
        print(f"  Max       : {s.max()}")
        print(f"  Mean      : {s.mean():.4f}")
        print(f"  Median    : {s.median():.4f}")
        print(f"  Q1        : {q1:.4f}")
        print(f"  Q3        : {q3:.4f}")
        print(f"  IQR       : {iqr:.4f}")
        print(f"  Outliers  : {len(outliers)}")
```

```python
for sheet, df in sheets_cleaned.items():
    if sheet == "Part V - Domestic Frauds":
        df["Report Year"] = df["Report Year"].astype("int64")
    elif "Year" in df.columns:
        df["Year"] = df["Year"].astype("int64")
```

```python
for sheet, df in sheets_cleaned.items():
    print(sheet)

    if sheet == "Part V - Domestic Frauds":
        print(df["Report Year"].dtype)
    else:
        print(df["Year"].dtype)
```

```python
# ============================================================
# FINAL DATA CLEANING VALIDATION
# ============================================================

for sheet, df in sheets_cleaned.items():

    print("\n" + "=" * 70)
    print(sheet)
    print("=" * 70)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))

    # Missing values
    print("Total missing values:", df.isna().sum().sum())

    # Duplicate rows
    print("Exact duplicate rows:", df.duplicated().sum())

    # Empty columns
    empty_cols = df.columns[df.isna().all()].tolist()
    print("Completely empty columns:", empty_cols)

    # Year validation
    if sheet == "Part V - Domestic Frauds":
        print("Report Year dtype:", df["Report Year"].dtype)
    else:
        print("Year dtype:", df["Year"].dtype)
```

```python
# ============================================================
# SAVE CLEANED DATA
# ============================================================

output_file = "PSI_Cleaned.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    for sheet, df in sheets_cleaned.items():
        df.to_excel(writer, sheet_name=sheet[:31], index=False)

print("Cleaned dataset saved successfully as:", output_file)
```
